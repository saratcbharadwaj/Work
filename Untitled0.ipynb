```python
import random
import math
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F

try:
    from IPython.display import display
except Exception:
    display = print


# ============================================================
# 1. CONFIGURATION & CHECKPOINT PATH
# ============================================================

class Config:
    ambient_dim = 5
    intrinsic_dims = [2, 3, 4, 5]
    families = ["Ellipsoid", "Hyperboloid", "Paraboloid"]
    num_classes = len(intrinsic_dims) * len(families)
    num_points = 256
    batch_size = 256
    seed = 42


CFG = Config()
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
OUT_DIR = Path(".")

SHEAR_LEVELS = [1, 5, 12, 20, 30]
TEST_CLOUDS_PER_CLASS = 100
TEST_NOISE = 0.10

ckpt_dir = OUT_DIR / "CPGMM" / "checkpoints"
if (ckpt_dir / "best_validations.pt").is_file():
    CPGMM_CKPT_PATH = ckpt_dir / "best_validations.pt"
elif (ckpt_dir / "best_validation.pt").is_file():
    CPGMM_CKPT_PATH = ckpt_dir / "best_validation.pt"
else:
    raise FileNotFoundError(f"Checkpoint not found in {ckpt_dir.resolve()}")

print(f"Device: {DEVICE} | Checkpoint: {CPGMM_CKPT_PATH}")


# ============================================================
# 2. 5D QUADRATIC OOD DATASET GENERATOR
# ============================================================

def random_rotation_matrix(dim):
    A = np.random.randn(dim, dim)
    Q, R = np.linalg.qr(A)
    signs = np.sign(np.diag(R))
    signs[signs == 0] = 1
    Q = Q @ np.diag(signs)
    if np.linalg.det(Q) < 0:
        Q[:, 0] *= -1
    return Q


def sample_intrinsic_geometry(intrinsic_dim, family, n_points):
    d = intrinsic_dim
    scales = np.random.uniform(0.7, 1.5, size=d)

    if family == "Ellipsoid":
        directions = np.random.randn(n_points, d)
        directions /= np.linalg.norm(directions, axis=1, keepdims=True) + 1e-12
        X = directions * scales
    elif family == "Hyperboloid":
        t = np.random.uniform(-1.5, 1.5, size=n_points)
        directions = np.random.randn(n_points, d - 1)
        directions /= np.linalg.norm(directions, axis=1, keepdims=True) + 1e-12
        radius = np.sqrt(1.0 + t ** 2)
        X = np.zeros((n_points, d), dtype=np.float64)
        X[:, :d - 1] = directions * radius[:, None]
        X[:, d - 1] = t
        X *= scales
    elif family == "Paraboloid":
        base = np.random.uniform(-0.8, 0.8, size=(n_points, d - 1))
        X = np.zeros((n_points, d), dtype=np.float64)
        X[:, :d - 1] = base
        X[:, d - 1] = np.sum(base ** 2, axis=1)
        X *= scales
    else:
        raise ValueError(f"Unknown family: {family}")

    return X


def embed_in_5d(X, intrinsic_dim):
    embedded = np.zeros((X.shape[0], CFG.ambient_dim), dtype=np.float64)
    embedded[:, :intrinsic_dim] = X
    return embedded


def apply_random_shear(X, shear_magnitude):
    D = X.shape[1]
    S = np.eye(D)
    for i in range(D):
        for j in range(i + 1, D):
            S[i, j] = np.random.uniform(-1.0, 1.0) * shear_magnitude
    return X @ S.T


def transform_cloud(X, shear_magnitude, noise_level):
    Q = random_rotation_matrix(CFG.ambient_dim)
    X = X @ Q.T
    X = apply_random_shear(X, shear_magnitude)
    X = X - X.mean(axis=0, keepdims=True)
    X += np.random.normal(0.0, noise_level, size=X.shape)
    rms = np.sqrt(np.mean(np.sum(X ** 2, axis=1))) + 1e-8
    X = X / rms
    return X.astype(np.float32)


def generate_ood_dataset(clouds_per_class, shear_magnitude, noise_level, seed):
    random.seed(seed)
    np.random.seed(seed)

    all_clouds, all_labels = [], []
    for dim_idx, intrinsic_dim in enumerate(CFG.intrinsic_dims):
        for family_idx, family in enumerate(CFG.families):
            class_id = dim_idx * len(CFG.families) + family_idx
            for _ in range(clouds_per_class):
                X_int = sample_intrinsic_geometry(intrinsic_dim, family, CFG.num_points)
                X_5d = embed_in_5d(X_int, intrinsic_dim)
                X_final = transform_cloud(X_5d, shear_magnitude, noise_level)
                all_clouds.append(X_final)
                all_labels.append(class_id)

    X = np.asarray(all_clouds, dtype=np.float32)
    y = np.asarray(all_labels, dtype=np.int64)
    idx = np.random.permutation(len(y))
    return X[idx], y[idx]


TEST_DATASETS = {}
for shear in SHEAR_LEVELS:
    X_test, y_test = generate_ood_dataset(
        clouds_per_class=TEST_CLOUDS_PER_CLASS,
        shear_magnitude=shear,
        noise_level=TEST_NOISE,
        seed=42 + shear
    )
    TEST_DATASETS[shear] = (
        torch.from_numpy(X_test).to(DEVICE),
        torch.from_numpy(y_test).to(DEVICE)
    )


def iterate_batches(X, y, batch_size):
    for start in range(0, X.shape[0], batch_size):
        yield X[start:start + batch_size], y[start:start + batch_size]


# ============================================================
# 3. CP-GMM ARCHITECTURE (FROM YOUR TRAINING CODE)
# ============================================================

def quadratic_veronese(X):
    B, N, D = X.shape
    terms = []
    for i in range(D):
        for j in range(i, D):
            terms.append(X[:, :, i] * X[:, :, j])
    for i in range(D):
        terms.append(X[:, :, i])
    terms.append(torch.ones_like(X[:, :, 0]))
    return torch.stack(terms, dim=-1)


def coefficient_matrix_from_vector(c):
    B = c.shape[0]
    A = torch.zeros(B, 5, 5, device=c.device, dtype=c.dtype)
    index = 0
    for i in range(5):
        for j in range(i, 5):
            value = c[:, index]
            if i == j:
                A[:, i, j] = value
            else:
                A[:, i, j] = value / 2.0
                A[:, j, i] = value / 2.0
            index += 1
    return A


class CPAttentionBlock(nn.Module):
    def __init__(self, dim=128, heads=8):
        super().__init__()
        self.dim = dim
        self.heads = heads
        self.head_dim = dim // heads
        self.norm1 = nn.LayerNorm(dim)
        self.qkv = nn.Linear(dim, dim * 3)
        self.projection = nn.Linear(dim, dim)
        self.norm2 = nn.LayerNorm(dim)
        self.ffn = nn.Sequential(
            nn.Linear(dim, dim * 4), nn.GELU(), nn.Linear(dim * 4, dim)
        )

    def forward(self, X, geometric_bias, alpha):
        B, N, D = X.shape
        Z = self.norm1(X)
        qkv = self.qkv(Z).reshape(B, N, 3, self.heads, self.head_dim)
        q, k, v = qkv.permute(2, 0, 3, 1, 4)

        attention = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_dim)
        attention = attention + alpha * geometric_bias[:, None, :, :]
        attention = torch.softmax(attention, dim=-1)

        output = (attention @ v).transpose(1, 2).reshape(B, N, D)
        X = X + self.projection(output)
        X = X + self.ffn(self.norm2(X))
        return X


class CPGMMNetwork(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.num_classes = num_classes
        self.embedding = nn.Linear(5 + 21, 128)
        self.router = nn.Sequential(
            nn.Linear(42, 128), nn.LayerNorm(128), nn.GELU(),
            nn.Linear(128, 64), nn.GELU(),
        )
        self.templates = nn.Parameter(torch.randn(num_classes, 64) * 0.02)
        self.router_temperature = 1.0
        self.template_projection = nn.Linear(64, 128)
        self.blocks = nn.ModuleList([
            CPAttentionBlock(128, 8),
            CPAttentionBlock(128, 8),
            CPAttentionBlock(128, 8),
        ])
        self.alpha = nn.Parameter(torch.tensor(1e-3))
        self.norm = nn.LayerNorm(128)
        self.classifier = nn.Sequential(
            nn.Linear(128, 256), nn.GELU(), nn.Dropout(0.3),
            nn.Linear(256, num_classes),
        )

    def algebraic_features(self, X):
        with torch.no_grad():
            lifted = quadratic_veronese(X)
            L64 = lifted.double()
            gram = L64.transpose(1, 2) @ L64
            eigvals, eigvecs = torch.linalg.eigh(gram)

            coefficients = eigvecs[:, :, 0]
            singular_values = torch.sqrt(
                torch.clamp(eigvals.flip(dims=[1]), min=0.0)
            )

            coefficients = coefficients.float()
            singular_values = singular_values.float()

            sign = torch.where(
                coefficients[:, 0:1] < 0,
                -torch.ones_like(coefficients[:, 0:1]),
                torch.ones_like(coefficients[:, 0:1]),
            )
            coefficients = coefficients * sign
            singular_values = singular_values / (
                singular_values.norm(dim=1, keepdim=True) + 1e-8
            )
            algebraic_input = torch.cat([coefficients, singular_values], dim=1)

        return lifted, coefficients, algebraic_input

    def forward(self, X):
        lifted, coefficients, algebraic_input = self.algebraic_features(X)

        router_features = self.router(algebraic_input)
        normalized_router = F.normalize(router_features, dim=1)
        normalized_templates = F.normalize(self.templates, dim=1)

        routing_logits = (
            normalized_router @ normalized_templates.T
        ) / self.router_temperature
        routing_weights = torch.softmax(routing_logits, dim=-1)

        routed_template = routing_weights @ self.templates
        routed_embedding = self.template_projection(routed_template)

        A = coefficient_matrix_from_vector(coefficients)
        geometric_bias = (X @ A @ X.transpose(1, 2)) / math.sqrt(21)

        tokens = torch.cat([X, lifted], dim=-1)
        tokens = self.embedding(tokens)
        tokens = tokens + routed_embedding[:, None, :]

        for block in self.blocks:
            tokens = block(tokens, geometric_bias, self.alpha)

        pooled = self.norm(tokens.mean(dim=1))
        logits = self.classifier(pooled)
        return logits, routing_weights


# ============================================================
# 4. ABLATION VARIANTS & EVALUATION
# ============================================================

def build_ablated_cpgmm(base_checkpoint, variant):
    model = CPGMMNetwork(CFG.num_classes).to(DEVICE)
    model.load_state_dict(base_checkpoint["model_state_dict"])
    model.eval()

    with torch.no_grad():
        if variant == "1_Full_CPGMM":
            pass
        elif variant == "2_No_Geometric_Bias":
            model.alpha.zero_()
        elif variant == "3_No_Template_Routing":
            model.template_projection.weight.zero_()
            model.template_projection.bias.zero_()
        elif variant == "4_No_Veronese_Tokens":
            model.embedding.weight[:, 5:].zero_()
        elif variant == "5_No_Algebraic_Branch":
            model.alpha.zero_()
            model.template_projection.weight.zero_()
            model.template_projection.bias.zero_()
        else:
            raise ValueError(f"Unknown variant: {variant}")

    return model


@torch.inference_mode()
def evaluate_accuracy(model, X, y):
    model.eval()
    correct, total = 0, 0
    for xb, yb in iterate_batches(X, y, CFG.batch_size):
        logits, _ = model(xb)
        correct += (logits.argmax(dim=1) == yb).sum().item()
        total += yb.size(0)
    return correct / total


ABLATION_VARIANTS = [
    "1_Full_CPGMM",
    "2_No_Geometric_Bias",
    "3_No_Template_Routing",
    "4_No_Veronese_Tokens",
    "5_No_Algebraic_Branch",
]

cpgmm_ckpt = torch.load(CPGMM_CKPT_PATH, map_location=DEVICE, weights_only=False)
ablation_records = []

for variant in ABLATION_VARIANTS:
    model = build_ablated_cpgmm(cpgmm_ckpt, variant)
    for shear in SHEAR_LEVELS:
        X_test, y_test = TEST_DATASETS[shear]
        acc = evaluate_accuracy(model, X_test, y_test)
        ablation_records.append({
            "Variant": variant,
            "Shear": shear,
            "Accuracy": acc * 100.0
        })
        print(f"{variant:<25} | Shear {shear:2d} | Accuracy: {acc:.2%}")
    del model
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

# ============================================================
# 5. SUMMARY TABLES & PLOT
# ============================================================

ablation_dir = OUT_DIR / "cpgmm_ablation"
ablation_dir.mkdir(parents=True, exist_ok=True)

ablation_df = pd.DataFrame(ablation_records)
ablation_table = ablation_df.pivot(index="Variant", columns="Shear", values="Accuracy")
ablation_table.columns = [f"Shear {s}" for s in ablation_table.columns]
ablation_table["Mean OOD Acc (%)"] = ablation_table.mean(axis=1)

delta_table = ablation_table.sub(ablation_table.loc["1_Full_CPGMM"], axis=1)

ablation_table.to_csv(ablation_dir / "cpgmm_ablation_accuracy.csv")
delta_table.to_csv(ablation_dir / "cpgmm_ablation_delta.csv")

print("\nCP-GMM ABLATION STUDY — ACCURACY (%):")
display(ablation_table.round(2))

print("\nACCURACY DROP RELATIVE TO FULL CP-GMM (Percentage Points):")
display(delta_table.round(2))

fig, ax = plt.subplots(figsize=(9, 5.5))
for variant in ABLATION_VARIANTS:
    sub = ablation_df[ablation_df["Variant"] == variant]
    ax.plot(sub["Shear"], sub["Accuracy"], marker="o", linewidth=2, label=variant)

ax.set_xlabel("Shear Magnitude")
ax.set_ylabel("12-Class Accuracy (%)")
ax.set_title("CP-GMM Inference-Time Ablation Study Across OOD Shear Levels")
ax.set_xticks(SHEAR_LEVELS)
ax.grid(True, linestyle="--", alpha=0.5)
ax.legend()
fig.tight_layout()
fig.savefig(ablation_dir / "cpgmm_ablation_curves.png", dpi=250)
plt.show()
plt.close(fig)

print(f"\nSaved to: {ablation_dir.resolve()}")
```

```python
import random
import math
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F

try:
    from IPython.display import display
except Exception:
    display = print


# ============================================================
# 1. CONFIGURATION & CHECKPOINTS
# ============================================================

class Config:
    ambient_dim = 5
    intrinsic_dims = [2, 3, 4, 5]
    families = ["Ellipsoid", "Hyperboloid", "Paraboloid"]
    num_classes = len(intrinsic_dims) * len(families)
    num_points = 256
    batch_size = 256
    seed = 42
    model_names = ["PointNet", "VanillaTransformer", "CPGMM"]


CFG = Config()
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
OUT_DIR = Path(".")

CLASS_NAMES = [f"{d}D_{family}" for d in CFG.intrinsic_dims for family in CFG.families]
DIMENSIONS = ["2D", "3D", "4D", "5D"]

SHEAR_LEVELS = [1, 5, 12, 20, 30]
TEST_CLOUDS_PER_CLASS = 100
TEST_NOISE = 0.10

PAPER_DIR = OUT_DIR / "paper_dimension_study"
PAPER_DIR.mkdir(parents=True, exist_ok=True)

CHECKPOINTS = {}
for name in CFG.model_names:
    ckpt_dir = OUT_DIR / name / "checkpoints"
    if (ckpt_dir / "best_validations.pt").is_file():
        CHECKPOINTS[name] = ckpt_dir / "best_validations.pt"
    elif (ckpt_dir / "best_validation.pt").is_file():
        CHECKPOINTS[name] = ckpt_dir / "best_validation.pt"
    else:
        raise FileNotFoundError(f"Missing checkpoint in {ckpt_dir.resolve()}")


# ============================================================
# 2. DATA GENERATOR
# ============================================================

def random_rotation_matrix(dim):
    A = np.random.randn(dim, dim)
    Q, R = np.linalg.qr(A)
    signs = np.sign(np.diag(R))
    signs[signs == 0] = 1
    Q = Q @ np.diag(signs)
    if np.linalg.det(Q) < 0:
        Q[:, 0] *= -1
    return Q


def sample_intrinsic_geometry(intrinsic_dim, family, n_points):
    d = intrinsic_dim
    scales = np.random.uniform(0.7, 1.5, size=d)

    if family == "Ellipsoid":
        directions = np.random.randn(n_points, d)
        directions /= np.linalg.norm(directions, axis=1, keepdims=True) + 1e-12
        X = directions * scales
    elif family == "Hyperboloid":
        t = np.random.uniform(-1.5, 1.5, size=n_points)
        directions = np.random.randn(n_points, d - 1)
        directions /= np.linalg.norm(directions, axis=1, keepdims=True) + 1e-12
        radius = np.sqrt(1.0 + t ** 2)
        X = np.zeros((n_points, d), dtype=np.float64)
        X[:, :d - 1] = directions * radius[:, None]
        X[:, d - 1] = t
        X *= scales
    elif family == "Paraboloid":
        base = np.random.uniform(-0.8, 0.8, size=(n_points, d - 1))
        X = np.zeros((n_points, d), dtype=np.float64)
        X[:, :d - 1] = base
        X[:, d - 1] = np.sum(base ** 2, axis=1)
        X *= scales
    else:
        raise ValueError(f"Unknown family: {family}")
    return X


def embed_in_5d(X, intrinsic_dim):
    embedded = np.zeros((X.shape[0], CFG.ambient_dim), dtype=np.float64)
    embedded[:, :intrinsic_dim] = X
    return embedded


def apply_random_shear(X, shear_magnitude):
    D = X.shape[1]
    S = np.eye(D)
    for i in range(D):
        for j in range(i + 1, D):
            S[i, j] = np.random.uniform(-1.0, 1.0) * shear_magnitude
    return X @ S.T


def transform_cloud(X, shear_magnitude, noise_level):
    Q = random_rotation_matrix(CFG.ambient_dim)
    X = X @ Q.T
    X = apply_random_shear(X, shear_magnitude)
    X = X - X.mean(axis=0, keepdims=True)
    X += np.random.normal(0.0, noise_level, size=X.shape)
    rms = np.sqrt(np.mean(np.sum(X ** 2, axis=1))) + 1e-8
    X = X / rms
    return X.astype(np.float32)


def generate_ood_dataset(clouds_per_class, shear_magnitude, noise_level, seed):
    random.seed(seed)
    np.random.seed(seed)

    all_clouds, all_labels = [], []
    for dim_idx, intrinsic_dim in enumerate(CFG.intrinsic_dims):
        for family_idx, family in enumerate(CFG.families):
            class_id = dim_idx * len(CFG.families) + family_idx
            for _ in range(clouds_per_class):
                X_int = sample_intrinsic_geometry(intrinsic_dim, family, CFG.num_points)
                X_5d = embed_in_5d(X_int, intrinsic_dim)
                X_final = transform_cloud(X_5d, shear_magnitude, noise_level)
                all_clouds.append(X_final)
                all_labels.append(class_id)

    X = np.asarray(all_clouds, dtype=np.float32)
    y = np.asarray(all_labels, dtype=np.int64)
    idx = np.random.permutation(len(y))
    return X[idx], y[idx]


TEST_DATASETS = {}
for shear in SHEAR_LEVELS:
    X_test, y_test = generate_ood_dataset(
        clouds_per_class=TEST_CLOUDS_PER_CLASS,
        shear_magnitude=shear,
        noise_level=TEST_NOISE,
        seed=42 + shear
    )
    TEST_DATASETS[shear] = (
        torch.from_numpy(X_test).to(DEVICE),
        torch.from_numpy(y_test).to(DEVICE)
    )


def iterate_batches(X, y, batch_size):
    for start in range(0, X.shape[0], batch_size):
        yield X[start:start + batch_size], y[start:start + batch_size]


# ============================================================
# 3. MODEL ARCHITECTURES
# ============================================================

class PointNetClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.point_mlp = nn.Sequential(
            nn.Conv1d(5, 64, 1), nn.BatchNorm1d(64), nn.ReLU(),
            nn.Conv1d(64, 128, 1), nn.BatchNorm1d(128), nn.ReLU(),
            nn.Conv1d(128, 256, 1), nn.BatchNorm1d(256), nn.ReLU(),
            nn.Conv1d(256, 512, 1), nn.BatchNorm1d(512), nn.ReLU(),
        )
        self.classifier = nn.Sequential(
            nn.Linear(512, 256), nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(256, 128), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(128, num_classes),
        )

    def forward(self, X):
        X = X.transpose(1, 2)
        features = self.point_mlp(X)
        pooled = torch.max(features, dim=2)[0]
        return self.classifier(pooled)


class TransformerBlock(nn.Module):
    def __init__(self, dim=128, heads=8):
        super().__init__()
        self.norm1 = nn.LayerNorm(dim)
        self.attention = nn.MultiheadAttention(embed_dim=dim, num_heads=heads, batch_first=True)
        self.norm2 = nn.LayerNorm(dim)
        self.ffn = nn.Sequential(
            nn.Linear(dim, dim * 4), nn.GELU(), nn.Linear(dim * 4, dim)
        )

    def forward(self, X):
        Z = self.norm1(X)
        attended, _ = self.attention(Z, Z, Z, need_weights=False)
        X = X + attended
        X = X + self.ffn(self.norm2(X))
        return X


class VanillaTransformer(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.embedding = nn.Linear(5, 128)
        self.blocks = nn.Sequential(
            TransformerBlock(128, 8),
            TransformerBlock(128, 8),
            TransformerBlock(128, 8),
        )
        self.norm = nn.LayerNorm(128)
        self.classifier = nn.Sequential(
            nn.Linear(128, 256), nn.GELU(), nn.Dropout(0.3),
            nn.Linear(256, num_classes),
        )

    def forward(self, X):
        X = self.embedding(X)
        X = self.blocks(X)
        X = self.norm(X)
        X = X.mean(dim=1)
        return self.classifier(X)


def quadratic_veronese(X):
    B, N, D = X.shape
    terms = []
    for i in range(D):
        for j in range(i, D):
            terms.append(X[:, :, i] * X[:, :, j])
    for i in range(D):
        terms.append(X[:, :, i])
    terms.append(torch.ones_like(X[:, :, 0]))
    return torch.stack(terms, dim=-1)


def coefficient_matrix_from_vector(c):
    B = c.shape[0]
    A = torch.zeros(B, 5, 5, device=c.device, dtype=c.dtype)
    index = 0
    for i in range(5):
        for j in range(i, 5):
            value = c[:, index]
            if i == j:
                A[:, i, j] = value
            else:
                A[:, i, j] = value / 2.0
                A[:, j, i] = value / 2.0
            index += 1
    return A


class CPAttentionBlock(nn.Module):
    def __init__(self, dim=128, heads=8):
        super().__init__()
        self.dim = dim
        self.heads = heads
        self.head_dim = dim // heads
        self.norm1 = nn.LayerNorm(dim)
        self.qkv = nn.Linear(dim, dim * 3)
        self.projection = nn.Linear(dim, dim)
        self.norm2 = nn.LayerNorm(dim)
        self.ffn = nn.Sequential(
            nn.Linear(dim, dim * 4), nn.GELU(), nn.Linear(dim * 4, dim)
        )

    def forward(self, X, geometric_bias, alpha):
        B, N, D = X.shape
        Z = self.norm1(X)
        qkv = self.qkv(Z).reshape(B, N, 3, self.heads, self.head_dim)
        q, k, v = qkv.permute(2, 0, 3, 1, 4)

        attention = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_dim)
        attention = attention + alpha * geometric_bias[:, None, :, :]
        attention = torch.softmax(attention, dim=-1)

        output = (attention @ v).transpose(1, 2).reshape(B, N, D)
        X = X + self.projection(output)
        X = X + self.ffn(self.norm2(X))
        return X


class CPGMMNetwork(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.num_classes = num_classes
        self.embedding = nn.Linear(5 + 21, 128)
        self.router = nn.Sequential(
            nn.Linear(42, 128), nn.LayerNorm(128), nn.GELU(),
            nn.Linear(128, 64), nn.GELU(),
        )
        self.templates = nn.Parameter(torch.randn(num_classes, 64) * 0.02)
        self.router_temperature = 1.0
        self.template_projection = nn.Linear(64, 128)
        self.blocks = nn.ModuleList([
            CPAttentionBlock(128, 8),
            CPAttentionBlock(128, 8),
            CPAttentionBlock(128, 8),
        ])
        self.alpha = nn.Parameter(torch.tensor(1e-3))
        self.norm = nn.LayerNorm(128)
        self.classifier = nn.Sequential(
            nn.Linear(128, 256), nn.GELU(), nn.Dropout(0.3),
            nn.Linear(256, num_classes),
        )

    def algebraic_features(self, X):
        with torch.no_grad():
            lifted = quadratic_veronese(X)
            L64 = lifted.double()
            gram = L64.transpose(1, 2) @ L64
            eigvals, eigvecs = torch.linalg.eigh(gram)

            coefficients = eigvecs[:, :, 0]
            singular_values = torch.sqrt(
                torch.clamp(eigvals.flip(dims=[1]), min=0.0)
            )

            coefficients = coefficients.float()
            singular_values = singular_values.float()

            sign = torch.where(
                coefficients[:, 0:1] < 0,
                -torch.ones_like(coefficients[:, 0:1]),
                torch.ones_like(coefficients[:, 0:1]),
            )
            coefficients = coefficients * sign
            singular_values = singular_values / (
                singular_values.norm(dim=1, keepdim=True) + 1e-8
            )
            algebraic_input = torch.cat([coefficients, singular_values], dim=1)

        return lifted, coefficients, algebraic_input

    def forward(self, X):
        lifted, coefficients, algebraic_input = self.algebraic_features(X)

        router_features = self.router(algebraic_input)
        normalized_router = F.normalize(router_features, dim=1)
        normalized_templates = F.normalize(self.templates, dim=1)

        routing_logits = (
            normalized_router @ normalized_templates.T
        ) / self.router_temperature
        routing_weights = torch.softmax(routing_logits, dim=-1)

        routed_template = routing_weights @ self.templates
        routed_embedding = self.template_projection(routed_template)

        A = coefficient_matrix_from_vector(coefficients)
        geometric_bias = (X @ A @ X.transpose(1, 2)) / math.sqrt(21)

        tokens = torch.cat([X, lifted], dim=-1)
        tokens = self.embedding(tokens)
        tokens = tokens + routed_embedding[:, None, :]

        for block in self.blocks:
            tokens = block(tokens, geometric_bias, self.alpha)

        pooled = self.norm(tokens.mean(dim=1))
        logits = self.classifier(pooled)
        return logits, routing_weights


def build_model(name):
    if name == "PointNet":
        return PointNetClassifier(CFG.num_classes)
    if name == "VanillaTransformer":
        return VanillaTransformer(CFG.num_classes)
    if name == "CPGMM":
        return CPGMMNetwork(CFG.num_classes)
    raise ValueError(name)


def get_logits(output):
    return output[0] if isinstance(output, tuple) else output


# ============================================================
# 4. EVALUATE ALL 3 MODELS BY INTRINSIC DIMENSION
# ============================================================

@torch.inference_mode()
def get_predictions(model, X, y):
    model.eval()
    all_true, all_pred = [], []
    for xb, yb in iterate_batches(X, y, CFG.batch_size):
        logits = get_logits(model(xb))
        all_true.extend(yb.cpu().numpy())
        all_pred.extend(logits.argmax(dim=1).cpu().numpy())
    return np.asarray(all_true), np.asarray(all_pred)


records = []

for model_name in CFG.model_names:
    ckpt = torch.load(CHECKPOINTS[model_name], map_location=DEVICE, weights_only=False)
    model = build_model(model_name).to(DEVICE)
    model.load_state_dict(ckpt["model_state_dict"])
    model.eval()

    for shear in SHEAR_LEVELS:
        X_test, y_test = TEST_DATASETS[shear]
        y_true, y_pred = get_predictions(model, X_test, y_test)

        # True and predicted intrinsic dimension index: 0->2D, 1->3D, 2->4D, 3->5D
        true_dim_idx = y_true // 3
        pred_dim_idx = y_pred // 3

        overall_12class = (y_true == y_pred).mean() * 100.0
        overall_dim_only = (true_dim_idx == pred_dim_idx).mean() * 100.0

        row = {
            "Model": model_name,
            "Shear": shear,
            "Overall_12Class": overall_12class,
            "Overall_DimOnly": overall_dim_only,
        }

        # Exact 12-class accuracy within each intrinsic dimension subset (2D, 3D, 4D, 5D)
        for dim_i, dim_label in enumerate(DIMENSIONS):
            mask = (true_dim_idx == dim_i)
            row[f"{dim_label}_Acc"] = (y_true[mask] == y_pred[mask]).mean() * 100.0
            row[f"{dim_label}_DimRecall"] = (pred_dim_idx[mask] == dim_i).mean() * 100.0

        records.append(row)

    del model, ckpt
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

df = pd.DataFrame(records)
df.to_csv(PAPER_DIR / "dimension_comparison_full.csv", index=False)


# ============================================================
# 5. PAPER-READY SUMMARY TABLES (MEAN ACROSS OOD SHEARS)
# ============================================================

mean_by_dim = (
    df.groupby("Model")[["2D_Acc", "3D_Acc", "4D_Acc", "5D_Acc", "Overall_12Class", "Overall_DimOnly"]]
    .mean()
    .reindex(CFG.model_names)
)
mean_by_dim.columns = ["2D (%)", "3D (%)", "4D (%)", "5D (%)", "12-Class Mean (%)", "Dim-Only Mean (%)"]
mean_by_dim.to_csv(PAPER_DIR / "paper_table_by_dimension.csv")

print("\nTABLE 1: MEAN OOD ACCURACY (%) ACROSS ALL SHEAR LEVELS BY INTRINSIC DIMENSION")
display(mean_by_dim.round(2))

# Also show per-shear breakdown for each dimension
print("\nTABLE 2: PER-SHEAR & PER-DIMENSION 12-CLASS ACCURACY (%)")
pivot_dim = df.pivot(index="Shear", columns="Model", values=["2D_Acc", "3D_Acc", "4D_Acc", "5D_Acc", "Overall_12Class"])
display(pivot_dim.round(2))

print("\n--- LaTeX Table Snippet for 4-Page Paper ---")
print(mean_by_dim.round(2).to_latex())


# ============================================================
# 6. COMPACT 2-PANEL PAPER FIGURE (FITS HALF-PAGE WIDTH)
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(11, 4.2))
colors = {"PointNet": "#1f77b4", "VanillaTransformer": "#ff7f0e", "CPGMM": "#2ca02c"}
markers = {"PointNet": "s", "VanillaTransformer": "^", "CPGMM": "o"}

# Panel A: Mean OOD Accuracy by Intrinsic Dimension (2D, 3D, 4D, 5D)
x = np.arange(len(DIMENSIONS))
width = 0.25
for idx, model_name in enumerate(CFG.model_names):
    vals = [mean_by_dim.loc[model_name, f"{d} (%)"] for d in DIMENSIONS]
    axes[0].bar(x + (idx - 1) * width, vals, width, label=model_name, color=colors[model_name])

axes[0].set_xticks(x)
axes[0].set_xticklabels(DIMENSIONS)
axes[0].set_xlabel("Intrinsic Manifold Dimension")
axes[0].set_ylabel("Mean OOD 12-Class Accuracy (%)")
axes[0].set_title("(a) Accuracy by Intrinsic Dimension")
axes[0].set_ylim(0, 100)
axes[0].grid(axis="y", linestyle="--", alpha=0.4)
axes[0].legend()

# Panel B: Overall OOD Accuracy vs. Shear Magnitude
for model_name in CFG.model_names:
    sub = df[df["Model"] == model_name]
    axes[1].plot(
        sub["Shear"], sub["Overall_12Class"],
        marker=markers[model_name], color=colors[model_name],
        linewidth=2, label=model_name
    )

axes[1].set_xlabel("Shear Magnitude")
axes[1].set_ylabel("Overall 12-Class Accuracy (%)")
axes[1].set_title("(b) OOD Shear Robustness")
axes[1].set_xticks(SHEAR_LEVELS)
axes[1].set_ylim(0, 100)
axes[1].grid(True, linestyle="--", alpha=0.4)
axes[1].legend()

fig.tight_layout()
fig.savefig(PAPER_DIR / "paper_dimension_comparison.pdf", bbox_inches="tight")
fig.savefig(PAPER_DIR / "paper_dimension_comparison.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close(fig)

print(f"\nSaved paper tables, LaTeX snippet, and vector PDF figure to: {PAPER_DIR.resolve()}")
```

```python
import random
import math
from pathlib import Path

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F

try:
    from IPython.display import display
except Exception:
    display = print


# ============================================================
# 1. CONFIGURATION & CHECKPOINTS
# ============================================================

class Config:
    ambient_dim = 5
    intrinsic_dims = [2, 3, 4, 5]
    families = ["Ellipsoid", "Hyperboloid", "Paraboloid"]
    num_classes = len(intrinsic_dims) * len(families)
    num_points = 256
    batch_size = 256
    seed = 42
    model_names = ["PointNet", "VanillaTransformer", "CPGMM"]


CFG = Config()
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
OUT_DIR = Path(".")

CLASS_NAMES = [f"{d}D_{family}" for d in CFG.intrinsic_dims for family in CFG.families]
DIMENSIONS = ["2D", "3D", "4D", "5D"]

SHEAR_LEVELS = [1, 5, 12, 20, 30]
TEST_CLOUDS_PER_CLASS = 100
TEST_NOISE = 0.10

FAIL_DIR = OUT_DIR / "paper_failure_analysis"
FAIL_DIR.mkdir(parents=True, exist_ok=True)

CHECKPOINTS = {}
for name in CFG.model_names:
    ckpt_dir = OUT_DIR / name / "checkpoints"
    if (ckpt_dir / "best_validations.pt").is_file():
        CHECKPOINTS[name] = ckpt_dir / "best_validations.pt"
    elif (ckpt_dir / "best_validation.pt").is_file():
        CHECKPOINTS[name] = ckpt_dir / "best_validation.pt"
    else:
        raise FileNotFoundError(f"Missing checkpoint in {ckpt_dir.resolve()}")


# ============================================================
# 2. DATA GENERATOR
# ============================================================

def random_rotation_matrix(dim):
    A = np.random.randn(dim, dim)
    Q, R = np.linalg.qr(A)
    signs = np.sign(np.diag(R))
    signs[signs == 0] = 1
    Q = Q @ np.diag(signs)
    if np.linalg.det(Q) < 0:
        Q[:, 0] *= -1
    return Q


def sample_intrinsic_geometry(intrinsic_dim, family, n_points):
    d = intrinsic_dim
    scales = np.random.uniform(0.7, 1.5, size=d)

    if family == "Ellipsoid":
        directions = np.random.randn(n_points, d)
        directions /= np.linalg.norm(directions, axis=1, keepdims=True) + 1e-12
        X = directions * scales
    elif family == "Hyperboloid":
        t = np.random.uniform(-1.5, 1.5, size=n_points)
        directions = np.random.randn(n_points, d - 1)
        directions /= np.linalg.norm(directions, axis=1, keepdims=True) + 1e-12
        radius = np.sqrt(1.0 + t ** 2)
        X = np.zeros((n_points, d), dtype=np.float64)
        X[:, :d - 1] = directions * radius[:, None]
        X[:, d - 1] = t
        X *= scales
    elif family == "Paraboloid":
        base = np.random.uniform(-0.8, 0.8, size=(n_points, d - 1))
        X = np.zeros((n_points, d), dtype=np.float64)
        X[:, :d - 1] = base
        X[:, d - 1] = np.sum(base ** 2, axis=1)
        X *= scales
    else:
        raise ValueError(f"Unknown family: {family}")
    return X


def embed_in_5d(X, intrinsic_dim):
    embedded = np.zeros((X.shape[0], CFG.ambient_dim), dtype=np.float64)
    embedded[:, :intrinsic_dim] = X
    return embedded


def apply_random_shear(X, shear_magnitude):
    D = X.shape[1]
    S = np.eye(D)
    for i in range(D):
        for j in range(i + 1, D):
            S[i, j] = np.random.uniform(-1.0, 1.0) * shear_magnitude
    return X @ S.T


def transform_cloud(X, shear_magnitude, noise_level):
    Q = random_rotation_matrix(CFG.ambient_dim)
    X = X @ Q.T
    X = apply_random_shear(X, shear_magnitude)
    X = X - X.mean(axis=0, keepdims=True)
    X += np.random.normal(0.0, noise_level, size=X.shape)
    rms = np.sqrt(np.mean(np.sum(X ** 2, axis=1))) + 1e-8
    X = X / rms
    return X.astype(np.float32)


def generate_ood_dataset(clouds_per_class, shear_magnitude, noise_level, seed):
    random.seed(seed)
    np.random.seed(seed)

    all_clouds, all_labels = [], []
    for dim_idx, intrinsic_dim in enumerate(CFG.intrinsic_dims):
        for family_idx, family in enumerate(CFG.families):
            class_id = dim_idx * len(CFG.families) + family_idx
            for _ in range(clouds_per_class):
                X_int = sample_intrinsic_geometry(intrinsic_dim, family, CFG.num_points)
                X_5d = embed_in_5d(X_int, intrinsic_dim)
                X_final = transform_cloud(X_5d, shear_magnitude, noise_level)
                all_clouds.append(X_final)
                all_labels.append(class_id)

    X = np.asarray(all_clouds, dtype=np.float32)
    y = np.asarray(all_labels, dtype=np.int64)
    idx = np.random.permutation(len(y))
    return X[idx], y[idx]


TEST_DATASETS = {}
for shear in SHEAR_LEVELS:
    X_test, y_test = generate_ood_dataset(
        clouds_per_class=TEST_CLOUDS_PER_CLASS,
        shear_magnitude=shear,
        noise_level=TEST_NOISE,
        seed=42 + shear
    )
    TEST_DATASETS[shear] = (
        torch.from_numpy(X_test).to(DEVICE),
        torch.from_numpy(y_test).to(DEVICE)
    )


def iterate_batches(X, y, batch_size):
    for start in range(0, X.shape[0], batch_size):
        yield X[start:start + batch_size], y[start:start + batch_size]


# ============================================================
# 3. MODEL ARCHITECTURES
# ============================================================

class PointNetClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.point_mlp = nn.Sequential(
            nn.Conv1d(5, 64, 1), nn.BatchNorm1d(64), nn.ReLU(),
            nn.Conv1d(64, 128, 1), nn.BatchNorm1d(128), nn.ReLU(),
            nn.Conv1d(128, 256, 1), nn.BatchNorm1d(256), nn.ReLU(),
            nn.Conv1d(256, 512, 1), nn.BatchNorm1d(512), nn.ReLU(),
        )
        self.classifier = nn.Sequential(
            nn.Linear(512, 256), nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(256, 128), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(128, num_classes),
        )

    def forward(self, X):
        X = X.transpose(1, 2)
        features = self.point_mlp(X)
        pooled = torch.max(features, dim=2)[0]
        return self.classifier(pooled)


class TransformerBlock(nn.Module):
    def __init__(self, dim=128, heads=8):
        super().__init__()
        self.norm1 = nn.LayerNorm(dim)
        self.attention = nn.MultiheadAttention(embed_dim=dim, num_heads=heads, batch_first=True)
        self.norm2 = nn.LayerNorm(dim)
        self.ffn = nn.Sequential(
            nn.Linear(dim, dim * 4), nn.GELU(), nn.Linear(dim * 4, dim)
        )

    def forward(self, X):
        Z = self.norm1(X)
        attended, _ = self.attention(Z, Z, Z, need_weights=False)
        X = X + attended
        X = X + self.ffn(self.norm2(X))
        return X


class VanillaTransformer(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.embedding = nn.Linear(5, 128)
        self.blocks = nn.Sequential(
            TransformerBlock(128, 8),
            TransformerBlock(128, 8),
            TransformerBlock(128, 8),
        )
        self.norm = nn.LayerNorm(128)
        self.classifier = nn.Sequential(
            nn.Linear(128, 256), nn.GELU(), nn.Dropout(0.3),
            nn.Linear(256, num_classes),
        )

    def forward(self, X):
        X = self.embedding(X)
        X = self.blocks(X)
        X = self.norm(X)
        X = X.mean(dim=1)
        return self.classifier(X)


def quadratic_veronese(X):
    B, N, D = X.shape
    terms = []
    for i in range(D):
        for j in range(i, D):
            terms.append(X[:, :, i] * X[:, :, j])
    for i in range(D):
        terms.append(X[:, :, i])
    terms.append(torch.ones_like(X[:, :, 0]))
    return torch.stack(terms, dim=-1)


def coefficient_matrix_from_vector(c):
    B = c.shape[0]
    A = torch.zeros(B, 5, 5, device=c.device, dtype=c.dtype)
    index = 0
    for i in range(5):
        for j in range(i, 5):
            value = c[:, index]
            if i == j:
                A[:, i, j] = value
            else:
                A[:, i, j] = value / 2.0
                A[:, j, i] = value / 2.0
            index += 1
    return A


class CPAttentionBlock(nn.Module):
    def __init__(self, dim=128, heads=8):
        super().__init__()
        self.dim = dim
        self.heads = heads
        self.head_dim = dim // heads
        self.norm1 = nn.LayerNorm(dim)
        self.qkv = nn.Linear(dim, dim * 3)
        self.projection = nn.Linear(dim, dim)
        self.norm2 = nn.LayerNorm(dim)
        self.ffn = nn.Sequential(
            nn.Linear(dim, dim * 4), nn.GELU(), nn.Linear(dim * 4, dim)
        )

    def forward(self, X, geometric_bias, alpha):
        B, N, D = X.shape
        Z = self.norm1(X)
        qkv = self.qkv(Z).reshape(B, N, 3, self.heads, self.head_dim)
        q, k, v = qkv.permute(2, 0, 3, 1, 4)

        attention = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_dim)
        attention = attention + alpha * geometric_bias[:, None, :, :]
        attention = torch.softmax(attention, dim=-1)

        output = (attention @ v).transpose(1, 2).reshape(B, N, D)
        X = X + self.projection(output)
        X = X + self.ffn(self.norm2(X))
        return X


class CPGMMNetwork(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.num_classes = num_classes
        self.embedding = nn.Linear(5 + 21, 128)
        self.router = nn.Sequential(
            nn.Linear(42, 128), nn.LayerNorm(128), nn.GELU(),
            nn.Linear(128, 64), nn.GELU(),
        )
        self.templates = nn.Parameter(torch.randn(num_classes, 64) * 0.02)
        self.router_temperature = 1.0
        self.template_projection = nn.Linear(64, 128)
        self.blocks = nn.ModuleList([
            CPAttentionBlock(128, 8),
            CPAttentionBlock(128, 8),
            CPAttentionBlock(128, 8),
        ])
        self.alpha = nn.Parameter(torch.tensor(1e-3))
        self.norm = nn.LayerNorm(128)
        self.classifier = nn.Sequential(
            nn.Linear(128, 256), nn.GELU(), nn.Dropout(0.3),
            nn.Linear(256, num_classes),
        )

    def algebraic_features(self, X):
        with torch.no_grad():
            lifted = quadratic_veronese(X)
            L64 = lifted.double()
            gram = L64.transpose(1, 2) @ L64
            eigvals, eigvecs = torch.linalg.eigh(gram)

            coefficients = eigvecs[:, :, 0]
            singular_values = torch.sqrt(
                torch.clamp(eigvals.flip(dims=[1]), min=0.0)
            )

            coefficients = coefficients.float()
            singular_values = singular_values.float()

            sign = torch.where(
                coefficients[:, 0:1] < 0,
                -torch.ones_like(coefficients[:, 0:1]),
                torch.ones_like(coefficients[:, 0:1]),
            )
            coefficients = coefficients * sign
            singular_values = singular_values / (
                singular_values.norm(dim=1, keepdim=True) + 1e-8
            )
            algebraic_input = torch.cat([coefficients, singular_values], dim=1)

        return lifted, coefficients, algebraic_input

    def forward(self, X):
        lifted, coefficients, algebraic_input = self.algebraic_features(X)

        router_features = self.router(algebraic_input)
        normalized_router = F.normalize(router_features, dim=1)
        normalized_templates = F.normalize(self.templates, dim=1)

        routing_logits = (
            normalized_router @ normalized_templates.T
        ) / self.router_temperature
        routing_weights = torch.softmax(routing_logits, dim=-1)

        routed_template = routing_weights @ self.templates
        routed_embedding = self.template_projection(routed_template)

        A = coefficient_matrix_from_vector(coefficients)
        geometric_bias = (X @ A @ X.transpose(1, 2)) / math.sqrt(21)

        tokens = torch.cat([X, lifted], dim=-1)
        tokens = self.embedding(tokens)
        tokens = tokens + routed_embedding[:, None, :]

        for block in self.blocks:
            tokens = block(tokens, geometric_bias, self.alpha)

        pooled = self.norm(tokens.mean(dim=1))
        logits = self.classifier(pooled)
        return logits, routing_weights


def build_model(name):
    if name == "PointNet":
        return PointNetClassifier(CFG.num_classes)
    if name == "VanillaTransformer":
        return VanillaTransformer(CFG.num_classes)
    if name == "CPGMM":
        return CPGMMNetwork(CFG.num_classes)
    raise ValueError(name)


def get_logits(output):
    return output[0] if isinstance(output, tuple) else output


# ============================================================
# 4. COLLECT AGGREGATED OOD CONFUSION MATRICES & FAILURE MODES
# ============================================================

@torch.inference_mode()
def get_predictions(model, X, y):
    model.eval()
    all_true, all_pred = [], []
    for xb, yb in iterate_batches(X, y, CFG.batch_size):
        logits = get_logits(model(xb))
        all_true.extend(yb.cpu().numpy())
        all_pred.extend(logits.argmax(dim=1).cpu().numpy())
    return np.asarray(all_true), np.asarray(all_pred)


agg_cm = {name: np.zeros((12, 12), dtype=np.int64) for name in CFG.model_names}
per_shear_class_acc = []

for model_name in CFG.model_names:
    ckpt = torch.load(CHECKPOINTS[model_name], map_location=DEVICE, weights_only=False)
    model = build_model(model_name).to(DEVICE)
    model.load_state_dict(ckpt["model_state_dict"])
    model.eval()

    for shear in SHEAR_LEVELS:
        X_test, y_test = TEST_DATASETS[shear]
        y_true, y_pred = get_predictions(model, X_test, y_test)
        np.add.at(agg_cm[model_name], (y_true, y_pred), 1)

        for cid, cname in enumerate(CLASS_NAMES):
            mask = (y_true == cid)
            acc = (y_pred[mask] == cid).mean() * 100.0
            dim_str, shape_str = cname.split("_")
            per_shear_class_acc.append({
                "Model": model_name,
                "Shear": shear,
                "Dimension": dim_str,
                "Shape": shape_str,
                "Class": cname,
                "Accuracy": acc
            })

    del model, ckpt
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

class_df = pd.DataFrame(per_shear_class_acc)


# ============================================================
# 5. TABLE A: ALL 12 CLASSES ACCURACY ACROSS MODELS
# ============================================================

mean_12_table = (
    class_df.pivot_table(index=["Dimension", "Shape"], columns="Model", values="Accuracy", aggfunc="mean")
    .reindex(index=[(d, s) for d in DIMENSIONS for s in CFG.families], columns=CFG.model_names)
)
mean_12_table.to_csv(FAIL_DIR / "all_12_classes_ood_accuracy.csv")

print("\n" + "="*75)
print("TABLE 1: MEAN OOD ACCURACY (%) FOR EVERY (DIMENSION, SHAPE) PAIR")
print("="*75)
display(mean_12_table.round(2))


# ============================================================
# 6. TABLE B: WORST-PERFORMING SHAPE & TOP MISCLASSIFICATION PER DIMENSION
# ============================================================

worst_rows = []

for dim_idx, dim_label in enumerate(DIMENSIONS):
    class_indices = [dim_idx * 3 + f_idx for f_idx in range(3)]  # 3 shapes in this dimension

    for model_name in CFG.model_names:
        cm = agg_cm[model_name]
        row_totals = cm.sum(axis=1)  # 500 samples per class across 5 shears

        # 1. Find which of the 3 shapes in this dimension has the lowest accuracy
        shape_accs = {}
        for cid in class_indices:
            shape_name = CLASS_NAMES[cid].split("_")[1]
            shape_accs[shape_name] = (cm[cid, cid] / row_totals[cid]) * 100.0

        worst_shape = min(shape_accs, key=shape_accs.get)
        worst_shape_acc = shape_accs[worst_shape]
        worst_cid = dim_idx * 3 + CFG.families.index(worst_shape)

        # 2. Where does that worst shape get misclassified most often?
        off_diag_worst = cm[worst_cid].copy()
        off_diag_worst[worst_cid] = -1
        top_wrong_for_worst_cid = int(np.argmax(off_diag_worst))
        top_wrong_for_worst_pct = (cm[worst_cid, top_wrong_for_worst_cid] / row_totals[worst_cid]) * 100.0

        # 3. Overall largest off-diagonal error pair in this dimension
        sub_cm = cm[class_indices, :].copy()
        for local_i, global_i in enumerate(class_indices):
            sub_cm[local_i, global_i] = -1
        flat_idx = int(np.argmax(sub_cm))
        local_true, global_pred = divmod(flat_idx, 12)
        global_true = class_indices[local_true]
        overall_pair_pct = (cm[global_true, global_pred] / row_totals[global_true]) * 100.0

        worst_rows.append({
            "Dimension": dim_label,
            "Model": model_name,
            "Worst Shape": worst_shape,
            "Worst Shape Acc (%)": round(worst_shape_acc, 2),
            "Worst Shape Confused With": f"{CLASS_NAMES[top_wrong_for_worst_cid]} ({top_wrong_for_worst_pct:.1f}%)",
            "All Shapes (E / H / P %)": f"{shape_accs['Ellipsoid']:.1f} / {shape_accs['Hyperboloid']:.1f} / {shape_accs['Paraboloid']:.1f}",
            "#1 Error Pair in Dim": f"{CLASS_NAMES[global_true]} -> {CLASS_NAMES[global_pred]} ({overall_pair_pct:.1f}%)"
        })

worst_df = pd.DataFrame(worst_rows)
worst_df.to_csv(FAIL_DIR / "worst_shape_per_dimension.csv", index=False)

print("\n" + "="*75)
print("TABLE 2: WORST SHAPE & PRIMARY FAILURE MODE BY DIMENSION")
print("="*75)
display(worst_df)


# ============================================================
# 7. TABLE C: CPGMM FAILURE PROGRESSION BY SHEAR ON WORST SHAPES
# ============================================================

cpgmm_df = class_df[class_df["Model"] == "CPGMM"]
cpgmm_pivot = cpgmm_df.pivot(index="Class", columns="Shear", values="Accuracy").reindex(CLASS_NAMES)
cpgmm_pivot["Mean (%)"] = cpgmm_pivot.mean(axis=1)
cpgmm_pivot.to_csv(FAIL_DIR / "cpgmm_per_class_by_shear.csv")

print("\n" + "="*75)
print("TABLE 3: CPGMM PER-CLASS ACCURACY (%) ACROSS SHEAR LEVELS")
print("="*75)
display(cpgmm_pivot.round(2))
print(f"\nSaved failure analysis CSVs to: {FAIL_DIR.resolve()}")
```

\section{Conclusion}

Because invertible affine transformations preserve a quadratic polynomial's algebraic form while severely distorting Euclidean distances, standard coordinate-based networks degrade under unseen shear. By lifting point clouds into a quadratic monomial space and routing their null-space invariants onto learned prototypes, CP-GMM conditions Transformer representations on explicit polynomial structure—with prototype routing directly driving the observed out-of-distribution gains. Because lower-dimensional quadratic zero-sets ($d \in \{2,3\}$) embedded in $\mathbb{R}^5$ satisfy multiple independent quadratic constraints, their null-space spectra remain stable under extreme shear, enabling CP-GMM to outperform both PointNet and Vanilla Transformer while preserving polynomial family identity. Conversely, because full-dimensional 5D quadratic surfaces rely on a single quadratic constraint, severe shear compresses minor axes into the noise floor, destabilizing coefficient recovery and causing 5D classes to collapse into 4D predictions. Stabilizing full-dimensional polynomial recovery via multi-vector subspace projections or shear-invariant normalization remains a clear next step.

```python
# ============================================================
# ONE-CELL PIPELINE
#   1. Locate checkpoints (searches folders, or auto-extracts *checkpoints*.zip)
#   2. Auto-train any MISSING seeds in TARGET_SEEDS (needs a GPU; resumable by re-running)
#   3. Evaluate every run on FIXED test sets (in-dist / noise-only / shear-only / paper-OOD)
#   4. CP-GMM inference-time ablations
#   5. Aggregate across seeds (mean +- std, paired CP-GMM minus baselines)
#
# Expected layout of each run:  <root>/<ModelName>/checkpoints/best_validation.pt
# Existing (already trained) roots are NEVER retrained or overwritten.
#
# Colab/Kaggle tip: upload ALL_checkpoints.zip (or the three <Model>_checkpoints.zip
# files) into the working directory; they are extracted automatically.
# ============================================================

import json
import math
import os
import random
import re
import time
import zipfile
from collections import defaultdict
from pathlib import Path

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F

# ------------------------- CONFIG -------------------------
CUDA = torch.cuda.is_available()
DEVICE = torch.device("cuda" if CUDA else "cpu")
WORK = Path("/kaggle/working") if Path("/kaggle/working").exists() else Path(".").resolve()

CKPT_ROOTS = []                 # optional: folders that contain PointNet/, VanillaTransformer/, CPGMM/
SEARCH_PATHS = [Path.cwd(), WORK, Path("/content"), Path("/kaggle/working"), Path("/kaggle/input")]
EXISTING_ROOT_SEED = 42         # seed assumed for a discovered root whose name has no "seedNN"
TARGET_SEEDS = [42, 43, 44]     # seeds to have in the final analysis
AUTO_TRAIN_MISSING = True       # train missing seeds automatically (GPU only)
ALLOW_CPU_TRAINING = False      # CPU training is impractically slow; keep False unless you must

MODEL_NAMES = ["PointNet", "VanillaTransformer", "CPGMM"]
CKPT_NAME = "best_validation.pt"

OUT = WORK / "eval_results"
OUT.mkdir(parents=True, exist_ok=True)

TEST_SEED = 2024                # must differ from every training seed
CLOUDS_PER_CLASS = 300          # per test cell
EVAL_BS = 200 if CUDA else 64
BOOTSTRAP = 1000
ABL_GROUPS = ("in-dist", "paper-OOD")   # cells used for ablations
ABL_PER_CLASS = 100                     # stratified subset per class for ablations

# training hyper-parameters (identical to the original notebook)
TRAIN_EPOCHS, TRAIN_LR, TRAIN_WD, TRAIN_BS = 3000, 1e-3, 1e-5, 256
ACC_THRESH, CHECK_EVERY, PATIENCE, RESUME_EVERY = 0.995, 50, 3, 50
TRAIN_PER_CLASS, VAL_PER_CLASS = 500, 100
TRAIN_SHEAR = (0.5, 2.5)
TRAIN_NOISE = 0.05

AMBIENT = 5
INTRINSIC_DIMS = [2, 3, 4, 5]
FAMILIES = ["Ellipsoid", "Hyperboloid", "Paraboloid"]
NUM_CLASSES = len(INTRINSIC_DIMS) * len(FAMILIES)
NUM_POINTS = 256
CLASS_NAMES = [f"{d}D_{f}" for d in INTRINSIC_DIMS for f in FAMILIES]

print("Device:", DEVICE, "| working dir:", WORK)

CELLS = (
    [("in-dist", "in-dist", ("uniform", 0.5, 2.5), 0.05),
     ("noise-only_n0.10", "noise-only", ("uniform", 0.5, 2.5), 0.10)]
    + [(f"shear-only_s{s}_n0.05", "shear-only", ("fixed", s), 0.05) for s in (5, 12, 20, 30)]
    + [(f"paper_s{s}_n0.10", "paper-OOD", ("fixed", s), 0.10) for s in (1, 5, 12, 20, 30)]
)


def seed_everything(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


# ------------------ DATA GENERATOR (identical to training notebook) ------------------
def random_rotation_matrix(dim):
    A = np.random.randn(dim, dim)
    Q, R = np.linalg.qr(A)
    signs = np.sign(np.diag(R))
    signs[signs == 0] = 1
    Q = Q @ np.diag(signs)
    if np.linalg.det(Q) < 0:
        Q[:, 0] *= -1
    return Q


def sample_intrinsic_geometry(d, family, n_points):
    scales = np.random.uniform(0.7, 1.5, size=d)
    if family == "Ellipsoid":
        directions = np.random.randn(n_points, d)
        directions /= np.linalg.norm(directions, axis=1, keepdims=True) + 1e-12
        X = directions * scales
    elif family == "Hyperboloid":
        t = np.random.uniform(-1.5, 1.5, size=n_points)
        directions = np.random.randn(n_points, d - 1)
        directions /= np.linalg.norm(directions, axis=1, keepdims=True) + 1e-12
        radius = np.sqrt(1.0 + t ** 2)
        X = np.zeros((n_points, d), dtype=np.float64)
        X[:, :d - 1] = directions * radius[:, None]
        X[:, d - 1] = t
        X *= scales
    elif family == "Paraboloid":
        base = np.random.uniform(-0.8, 0.8, size=(n_points, d - 1))
        X = np.zeros((n_points, d), dtype=np.float64)
        X[:, :d - 1] = base
        X[:, d - 1] = np.sum(base ** 2, axis=1)
        X *= scales
    else:
        raise ValueError(family)
    return X


def embed_in_5d(X, d):
    out = np.zeros((X.shape[0], AMBIENT), dtype=np.float64)
    out[:, :d] = X
    return out


def apply_random_shear(X, s):
    D = X.shape[1]
    S = np.eye(D)
    for i in range(D):
        for j in range(i + 1, D):
            S[i, j] = np.random.uniform(-1.0, 1.0) * s
    return X @ S.T


def transform_cloud(X, shear, noise):
    """Same ops/order as training; also returns the pre-normalisation RMS radius."""
    Q = random_rotation_matrix(AMBIENT)
    X = X @ Q.T
    X = apply_random_shear(X, shear)
    X = X - X.mean(axis=0, keepdims=True)
    X += np.random.normal(0.0, noise, size=X.shape)
    rms = np.sqrt(np.mean(np.sum(X ** 2, axis=1))) + 1e-8
    return (X / rms).astype(np.float32), float(rms)


def generate_cell(shear_spec, noise, seed, per_class):
    np.random.seed(seed)
    clouds, labels, rmss = [], [], []
    for di, d in enumerate(INTRINSIC_DIMS):
        for fi, fam in enumerate(FAMILIES):
            cid = di * len(FAMILIES) + fi
            for _ in range(per_class):
                X5 = embed_in_5d(sample_intrinsic_geometry(d, fam, NUM_POINTS), d)
                s = (np.random.uniform(shear_spec[1], shear_spec[2])
                     if shear_spec[0] == "uniform" else float(shear_spec[1]))
                Xf, rms = transform_cloud(X5, s, noise)
                clouds.append(Xf)
                labels.append(cid)
                rmss.append(rms)
    return (np.asarray(clouds, np.float32), np.asarray(labels, np.int64),
            np.asarray(rmss, np.float64))


def generate_train_split(per_class):
    """Exact analogue of the notebook's generate_dataset (uses the global RNG stream)."""
    clouds, labels = [], []
    for di, d in enumerate(INTRINSIC_DIMS):
        for fi, fam in enumerate(FAMILIES):
            cid = di * len(FAMILIES) + fi
            for _ in range(per_class):
                X5 = embed_in_5d(sample_intrinsic_geometry(d, fam, NUM_POINTS), d)
                s = np.random.uniform(TRAIN_SHEAR[0], TRAIN_SHEAR[1])
                Xf, _ = transform_cloud(X5, s, TRAIN_NOISE)
                clouds.append(Xf)
                labels.append(cid)
    X = np.asarray(clouds, np.float32)
    y = np.asarray(labels, np.int64)
    idx = np.random.permutation(len(y))
    return X[idx], y[idx]


def load_or_build_test_sets():
    cache = OUT / f"test_sets_seed{TEST_SEED}_n{CLOUDS_PER_CLASS}.npz"
    if cache.exists():
        print("Loading cached test sets:", cache)
        z = np.load(cache)
        return [(z[f"X{i}"], z[f"y{i}"], z[f"r{i}"]) for i in range(len(CELLS))]
    print("Generating test sets (one-off)...")
    sets, arrs = [], {}
    for i, (name, group, spec, noise) in enumerate(CELLS):
        X, y, r = generate_cell(spec, noise, TEST_SEED + 1000 * i, CLOUDS_PER_CLASS)
        sets.append((X, y, r))
        arrs[f"X{i}"], arrs[f"y{i}"], arrs[f"r{i}"] = X, y, r
    np.savez(cache, **arrs)
    return sets


# ------------------ MODELS (identical to training; attribute names must match) ------------------
class PointNetClassifier(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.point_mlp = nn.Sequential(
            nn.Conv1d(5, 64, 1), nn.BatchNorm1d(64), nn.ReLU(),
            nn.Conv1d(64, 128, 1), nn.BatchNorm1d(128), nn.ReLU(),
            nn.Conv1d(128, 256, 1), nn.BatchNorm1d(256), nn.ReLU(),
            nn.Conv1d(256, 512, 1), nn.BatchNorm1d(512), nn.ReLU(),
        )
        self.classifier = nn.Sequential(
            nn.Linear(512, 256), nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(256, 128), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(128, num_classes),
        )

    def forward(self, X):
        X = X.transpose(1, 2)
        features = self.point_mlp(X)
        pooled = torch.max(features, dim=2)[0]
        return self.classifier(pooled)


class TransformerBlock(nn.Module):
    def __init__(self, dim=128, heads=8):
        super().__init__()
        self.norm1 = nn.LayerNorm(dim)
        self.attention = nn.MultiheadAttention(embed_dim=dim, num_heads=heads, batch_first=True)
        self.norm2 = nn.LayerNorm(dim)
        self.ffn = nn.Sequential(nn.Linear(dim, dim * 4), nn.GELU(), nn.Linear(dim * 4, dim))

    def forward(self, X):
        Z = self.norm1(X)
        attended, _ = self.attention(Z, Z, Z, need_weights=False)
        X = X + attended
        X = X + self.ffn(self.norm2(X))
        return X


class VanillaTransformer(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.embedding = nn.Linear(5, 128)
        self.blocks = nn.Sequential(TransformerBlock(128, 8), TransformerBlock(128, 8),
                                    TransformerBlock(128, 8))
        self.norm = nn.LayerNorm(128)
        self.classifier = nn.Sequential(nn.Linear(128, 256), nn.GELU(), nn.Dropout(0.3),
                                        nn.Linear(256, num_classes))

    def forward(self, X):
        X = self.embedding(X)
        X = self.blocks(X)
        X = self.norm(X)
        X = X.mean(dim=1)
        return self.classifier(X)


def quadratic_veronese(X):
    B, N, D = X.shape
    terms = []
    for i in range(D):
        for j in range(i, D):
            terms.append(X[:, :, i] * X[:, :, j])
    for i in range(D):
        terms.append(X[:, :, i])
    terms.append(torch.ones_like(X[:, :, 0]))
    return torch.stack(terms, dim=-1)


def coefficient_matrix_from_vector(c):
    B = c.shape[0]
    A = torch.zeros(B, 5, 5, device=c.device, dtype=c.dtype)
    index = 0
    for i in range(5):
        for j in range(i, 5):
            value = c[:, index]
            if i == j:
                A[:, i, j] = value
            else:
                A[:, i, j] = value / 2.0
                A[:, j, i] = value / 2.0
            index += 1
    return A


class CPAttentionBlock(nn.Module):
    def __init__(self, dim=128, heads=8):
        super().__init__()
        self.dim, self.heads, self.head_dim = dim, heads, dim // heads
        self.norm1 = nn.LayerNorm(dim)
        self.qkv = nn.Linear(dim, dim * 3)
        self.projection = nn.Linear(dim, dim)
        self.norm2 = nn.LayerNorm(dim)
        self.ffn = nn.Sequential(nn.Linear(dim, dim * 4), nn.GELU(), nn.Linear(dim * 4, dim))

    def forward(self, X, geometric_bias, alpha):
        B, N, D = X.shape
        Z = self.norm1(X)
        qkv = self.qkv(Z).reshape(B, N, 3, self.heads, self.head_dim)
        q, k, v = qkv.permute(2, 0, 3, 1, 4)
        attention = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_dim)
        attention = attention + alpha * geometric_bias[:, None, :, :]
        attention = torch.softmax(attention, dim=-1)
        output = (attention @ v).transpose(1, 2).reshape(B, N, D)
        X = X + self.projection(output)
        X = X + self.ffn(self.norm2(X))
        return X


class CPGMMNetwork(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.num_classes = num_classes
        self.embedding = nn.Linear(5 + 21, 128)
        self.router = nn.Sequential(
            nn.Linear(42, 128), nn.LayerNorm(128), nn.GELU(),
            nn.Linear(128, 64), nn.GELU(),
        )
        self.templates = nn.Parameter(torch.randn(num_classes, 64) * 0.02)
        self.router_temperature = 1.0
        self.template_projection = nn.Linear(64, 128)
        self.blocks = nn.ModuleList([CPAttentionBlock(128, 8) for _ in range(3)])
        self.alpha = nn.Parameter(torch.tensor(1e-3))
        self.norm = nn.LayerNorm(128)
        self.classifier = nn.Sequential(nn.Linear(128, 256), nn.GELU(), nn.Dropout(0.3),
                                        nn.Linear(256, num_classes))

    def algebraic_features(self, X):
        with torch.no_grad():
            lifted = quadratic_veronese(X)
            L64 = lifted.double()
            gram = L64.transpose(1, 2) @ L64
            eigvals, eigvecs = torch.linalg.eigh(gram)
            coefficients = eigvecs[:, :, 0]
            singular_values = torch.sqrt(torch.clamp(eigvals.flip(dims=[1]), min=0.0))
            coefficients = coefficients.float()
            singular_values = singular_values.float()
            sign = torch.where(coefficients[:, 0:1] < 0,
                               -torch.ones_like(coefficients[:, 0:1]),
                               torch.ones_like(coefficients[:, 0:1]))
            coefficients = coefficients * sign
            singular_values = singular_values / (singular_values.norm(dim=1, keepdim=True) + 1e-8)
            algebraic_input = torch.cat([coefficients, singular_values], dim=1)
        return lifted, coefficients, algebraic_input

    def forward(self, X):
        lifted, coefficients, algebraic_input = self.algebraic_features(X)
        router_features = self.router(algebraic_input)
        normalized_router = F.normalize(router_features, dim=1)
        normalized_templates = F.normalize(self.templates, dim=1)
        routing_logits = (normalized_router @ normalized_templates.T) / self.router_temperature
        routing_weights = torch.softmax(routing_logits, dim=-1)
        routed_template = routing_weights @ self.templates
        routed_embedding = self.template_projection(routed_template)
        A = coefficient_matrix_from_vector(coefficients)
        geometric_bias = (X @ A @ X.transpose(1, 2)) / math.sqrt(21)
        tokens = torch.cat([X, lifted], dim=-1)
        tokens = self.embedding(tokens)
        tokens = tokens + routed_embedding[:, None, :]
        for block in self.blocks:
            tokens = block(tokens, geometric_bias, self.alpha)
        pooled = self.norm(tokens.mean(dim=1))
        return self.classifier(pooled), routing_weights


def build_model(name):
    return {"PointNet": PointNetClassifier,
            "VanillaTransformer": VanillaTransformer,
            "CPGMM": CPGMMNetwork}[name](NUM_CLASSES)


def logits_of(model, xb):
    out = model(xb)
    return out[0] if isinstance(out, tuple) else out


# ------------------ TRAINING (only used for missing seeds) ------------------
def atomic_save(obj, path):
    path = Path(path)
    tmp = path.with_suffix(path.suffix + ".tmp")
    torch.save(obj, tmp)
    os.replace(tmp, path)


def batches(X, y, bs, shuffle, gen=None):
    n = X.shape[0]
    order = torch.randperm(n, generator=gen).to(X.device) if shuffle \
        else torch.arange(n, device=X.device)
    for s in range(0, n, bs):
        idx = order[s:s + bs]
        yield X[idx], y[idx]


@torch.no_grad()
def eval_loop(model, X, y):
    model.eval()
    crit = nn.CrossEntropyLoss()
    tl, tc, tn = 0.0, 0, 0
    for xb, yb in batches(X, y, TRAIN_BS, False):
        lg = logits_of(model, xb)
        tl += crit(lg, yb).item() * yb.size(0)
        tc += (lg.argmax(1) == yb).sum().item()
        tn += yb.size(0)
    return tl / tn, tc / tn


def model_complete(mdir):
    ck = mdir / "checkpoints"
    return (ck / "final_epoch.pt").exists() or (mdir / "DONE.flag").exists() \
        or (mdir / "training_summary.json").exists()


def train_model(name, seed, root, data):
    Xtr, Ytr, Xva, Yva = data
    mdir = root / name
    ckdir = mdir / "checkpoints"
    ckdir.mkdir(parents=True, exist_ok=True)
    best_path, final_path, last_path = ckdir / CKPT_NAME, ckdir / "final_epoch.pt", ckdir / "last_resume.pt"
    print(f"\n[train] seed {seed} / {name}  ->  {mdir}")

    seed_everything(seed)
    model = build_model(name).to(DEVICE)
    opt = torch.optim.AdamW(model.parameters(), lr=TRAIN_LR, weight_decay=TRAIN_WD)
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=TRAIN_EPOCHS)
    crit = nn.CrossEntropyLoss()
    gen = torch.Generator()
    gen.manual_seed(seed + 100)
    history, state = [], {"best_val": -1.0, "best_val_epoch": -1, "streak": 0}
    start_epoch, elapsed_before = 1, 0.0

    if last_path.exists():
        ck = torch.load(last_path, map_location=DEVICE, weights_only=False)
        model.load_state_dict(ck["model_state_dict"])
        opt.load_state_dict(ck["optimizer_state_dict"])
        sched.load_state_dict(ck["scheduler_state_dict"])
        history, state, start_epoch = ck["history"], ck["state"], ck["epoch"] + 1
        elapsed_before = ck.get("elapsed", 0.0)
        gen.manual_seed(seed + 100 + start_epoch)
        print(f"  resuming from epoch {ck['epoch']}")

    t0 = time.perf_counter()
    early, epoch = False, start_epoch - 1
    for epoch in range(start_epoch, TRAIN_EPOCHS + 1):
        model.train()
        tl, tc, tn = 0.0, 0, 0
        for xb, yb in batches(Xtr, Ytr, TRAIN_BS, True, gen):
            opt.zero_grad(set_to_none=True)
            lg = logits_of(model, xb)
            loss = crit(lg, yb)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            opt.step()
            tl += loss.item() * yb.size(0)
            tc += (lg.argmax(1) == yb).sum().item()
            tn += yb.size(0)
        sched.step()
        tr_acc = tc / tn
        vl, va = eval_loop(model, Xva, Yva)
        history.append({"epoch": epoch, "train_loss": tl / tn, "train_accuracy": tr_acc,
                        "val_loss": vl, "val_accuracy": va})

        if va > state["best_val"]:
            state["best_val"], state["best_val_epoch"] = va, epoch
            atomic_save({"model_name": name, "epoch": epoch, "seed": seed,
                         "model_state_dict": model.state_dict(),
                         "best_val_accuracy": va, "best_val_epoch": epoch}, best_path)

        if epoch == 1 or epoch % CHECK_EVERY == 0:
            print(f"  ep {epoch:04d} | train {tr_acc:.4f} | val {va:.4f} | best val {state['best_val']:.4f}",
                  flush=True)
        if epoch % CHECK_EVERY == 0:
            state["streak"] = state["streak"] + 1 if tr_acc >= ACC_THRESH else 0
            if state["streak"] >= PATIENCE:
                early = True
                print(f"  early stopping at epoch {epoch}")
        if epoch % RESUME_EVERY == 0 and not early:
            atomic_save({"model_state_dict": model.state_dict(),
                         "optimizer_state_dict": opt.state_dict(),
                         "scheduler_state_dict": sched.state_dict(),
                         "epoch": epoch, "history": history, "state": state,
                         "elapsed": elapsed_before + time.perf_counter() - t0}, last_path)
        if early:
            break

    secs = elapsed_before + time.perf_counter() - t0
    atomic_save({"model_name": name, "epoch": epoch, "seed": seed,
                 "model_state_dict": model.state_dict()}, final_path)
    pd.DataFrame(history).to_csv(mdir / "training_history.csv", index=False)
    summary = {"model": name, "seed": seed, "actual_epochs": epoch,
               "early_stopping_triggered": early,
               "best_validation_accuracy": state["best_val"],
               "best_validation_epoch": state["best_val_epoch"],
               "training_time_seconds": secs, "seconds_per_epoch": secs / max(epoch, 1),
               "parameter_count": sum(p.numel() for p in model.parameters())}
    (mdir / "training_summary.json").write_text(json.dumps(summary, indent=2))
    (mdir / "DONE.flag").write_text("done")
    if last_path.exists():
        last_path.unlink()
    print(f"  done: {epoch} epochs, {secs / 60:.1f} min, best val {state['best_val']:.4f}")


def train_seed(seed, root):
    pending = [m for m in MODEL_NAMES if not model_complete(root / m)]
    if not pending:
        return
    root.mkdir(parents=True, exist_ok=True)
    print(f"\n=== Auto-training seed {seed} -> {root} (pending: {pending}) ===")
    seed_everything(seed)                       # same order as the notebook:
    Xtr, ytr = generate_train_split(TRAIN_PER_CLASS)   # train split first,
    Xva, yva = generate_train_split(VAL_PER_CLASS)     # then validation split
    data = (torch.from_numpy(Xtr).to(DEVICE), torch.from_numpy(ytr).to(DEVICE),
            torch.from_numpy(Xva).to(DEVICE), torch.from_numpy(yva).to(DEVICE))
    for name in MODEL_NAMES:
        if name in pending:
            train_model(name, seed, root, data)


# ------------------ CHECKPOINT DISCOVERY ------------------
def _model_key(n):
    return {m.lower(): m for m in MODEL_NAMES}.get(n.lower())


def find_files(base, fname, max_depth=6):
    base, out = Path(base), []
    if not base.exists():
        return out
    base_depth = len(base.parts)
    for dirpath, dirnames, filenames in os.walk(base):
        depth = len(Path(dirpath).parts) - base_depth
        dirnames[:] = [] if depth >= max_depth else [d for d in dirnames if not d.startswith(".")]
        if fname in filenames:
            out.append(Path(dirpath) / fname)
    return out


def discover_roots():
    roots = defaultdict(set)
    for r in CKPT_ROOTS:
        for m in MODEL_NAMES:
            if (Path(r) / m / "checkpoints" / CKPT_NAME).exists():
                roots[Path(r).resolve()].add(m)
    seen = set()
    for base in SEARCH_PATHS:
        base = Path(base)
        if base in seen or not base.exists():
            continue
        seen.add(base)
        for p in find_files(base, CKPT_NAME):
            if p.parent.name != "checkpoints":
                continue
            mdir = p.parent.parent
            mk = _model_key(mdir.name)
            if mk:
                roots[mdir.parent.resolve()].add(mk)
    return roots


def restore_from_zips():
    zips = set()
    for base in SEARCH_PATHS:
        base = Path(base)
        if base.exists():
            zips |= {p for p in base.glob("*checkpoints*.zip")}
    if not zips:
        return False
    dest = WORK / "restored_checkpoints"
    dest.mkdir(exist_ok=True)
    for z in sorted(zips):
        try:
            with zipfile.ZipFile(z) as zf:
                zf.extractall(dest)
            print("Extracted", z)
        except Exception as e:
            print("Could not extract", z, e)
    return True


def seed_of(root):
    m = re.search(r"seed[_-]?(\d+)", root.name, re.I)
    return int(m.group(1)) if m else EXISTING_ROOT_SEED


def build_run_table():
    roots = discover_roots()
    if not roots and restore_from_zips():
        roots = discover_roots()
    runs = {}
    for root, models in sorted(roots.items(), key=lambda kv: -len(kv[1])):
        label = f"seed{seed_of(root)}"
        if label not in runs:
            runs[label] = root
    return runs


# ------------------ CP-GMM FORWARD WITH INFERENCE-TIME ABLATIONS ------------------
@torch.no_grad()
def cpgmm_logits(model, X, mode="full", X_sig=None, const_emb=None, return_emb=False):
    """Replica of CPGMMNetwork.forward with switches.
    mode: full | no_bias | no_proto | const_proto | zero_c | zero_sigma | shuffle_sig
    X_sig: clouds the signature is computed from (shuffle_sig)."""
    src = X_sig if X_sig is not None else X
    _, coef, alg = model.algebraic_features(src)
    lifted = quadratic_veronese(X)
    if mode == "zero_c":
        alg = alg.clone()
        alg[:, :21] = 0
        coef = torch.zeros_like(coef)
    elif mode == "zero_sigma":
        alg = alg.clone()
        alg[:, 21:] = 0
    rf = model.router(alg)
    nr = F.normalize(rf, dim=1)
    nt = F.normalize(model.templates, dim=1)
    w = torch.softmax((nr @ nt.T) / model.router_temperature, dim=-1)
    emb = model.template_projection(w @ model.templates)
    if mode == "no_proto":
        emb = torch.zeros_like(emb)
    elif mode == "const_proto":
        emb = const_emb.expand_as(emb)
    A = coefficient_matrix_from_vector(coef)
    bias = (X @ A @ X.transpose(1, 2)) / math.sqrt(21)
    alpha = torch.zeros_like(model.alpha) if mode == "no_bias" else model.alpha
    tokens = model.embedding(torch.cat([X, lifted], dim=-1)) + emb[:, None, :]
    for blk in model.blocks:
        tokens = blk(tokens, bias, alpha)
    logits = model.classifier(model.norm(tokens.mean(dim=1)))
    return (logits, emb) if return_emb else logits


@torch.no_grad()
def predict(model, name, X_np, mode="full", src_idx=None, const_emb=None):
    X = torch.from_numpy(X_np)
    preds = []
    for s in range(0, X.shape[0], EVAL_BS):
        xb = X[s:s + EVAL_BS].to(DEVICE)
        if name == "CPGMM":
            xs = None if src_idx is None else X[src_idx[s:s + EVAL_BS]].to(DEVICE)
            lg = cpgmm_logits(model, xb, mode, xs, const_emb)
        else:
            lg = model(xb)
        preds.append(lg.argmax(1).cpu())
    return torch.cat(preds).numpy()


@torch.no_grad()
def mean_proto_embedding(model, X_np):
    X = torch.from_numpy(X_np)
    embs = []
    for s in range(0, X.shape[0], EVAL_BS):
        _, e = cpgmm_logits(model, X[s:s + EVAL_BS].to(DEVICE), "full", return_emb=True)
        embs.append(e.cpu())
    return torch.cat(embs).mean(0, keepdim=True).to(DEVICE)


def mismatched_source_idx(y, seed=0):
    """Index of a cloud from a DIFFERENT class for every cloud (signature donor)."""
    g = torch.Generator().manual_seed(seed)
    yt = torch.from_numpy(y)
    idx = torch.randperm(len(y), generator=g)
    for _ in range(200):
        bad = (yt[idx] == yt).nonzero().flatten()
        if len(bad) == 0:
            break
        idx[bad] = torch.randint(0, len(y), (len(bad),), generator=g)
    return idx


def stratified_idx(y, per_class, seed):
    rng = np.random.default_rng(seed)
    keep = []
    for c in range(NUM_CLASSES):
        ids = np.where(y == c)[0]
        keep.append(rng.permutation(ids)[:per_class])
    return np.sort(np.concatenate(keep))


# ------------------ METRICS ------------------
def boot_ci(correct, B=BOOTSTRAP, seed=0):
    rng = np.random.default_rng(seed)
    n = len(correct)
    idx = rng.integers(0, n, size=(B, n))
    means = correct.astype(np.float32)[idx].mean(axis=1)
    return float(np.percentile(means, 2.5)), float(np.percentile(means, 97.5))


def metrics(pred, y):
    correct = pred == y
    d, fam = y // len(FAMILIES), y % len(FAMILIES)
    out = {"acc12": correct.mean() * 100,
           "acc_dim": ((pred // len(FAMILIES)) == d).mean() * 100,
           "acc_family": ((pred % len(FAMILIES)) == fam).mean() * 100}
    for i, dim in enumerate(INTRINSIC_DIMS):
        out[f"acc_{dim}D"] = correct[d == i].mean() * 100
    return out, correct


def load_model(path, name):
    ck = torch.load(path, map_location=DEVICE, weights_only=False)
    model = build_model(name).to(DEVICE)
    model.load_state_dict(ck["model_state_dict"])
    model.eval()
    return model, ck


def pm(x):
    x = np.asarray(x, dtype=float)
    if len(x) > 1:
        return f"{x.mean():.2f} ± {x.std(ddof=1):.2f} (n={len(x)})"
    return f"{x.mean():.2f} (n=1)"


# ============================================================
# MAIN
# ============================================================
t_start = time.time()

# ---- 1. locate runs ----
runs = build_run_table()
print("\nRuns found:", {k: str(v) for k, v in runs.items()} or "none")

# ---- 2. auto-train missing seeds ----
base_dir = next(iter(runs.values())).parent if runs else WORK
can_train = AUTO_TRAIN_MISSING and (CUDA or ALLOW_CPU_TRAINING)
if AUTO_TRAIN_MISSING and not can_train:
    print("\n[info] Auto-training skipped: no GPU. Switch the runtime to a GPU "
          "(or set ALLOW_CPU_TRAINING=True) to train the missing seeds automatically.")
if can_train:
    for s in TARGET_SEEDS:
        label = f"seed{s}"
        new_root = (base_dir / f"quadratic_5D_experiment_seed{s}").resolve()
        if label in runs and runs[label] != new_root:
            continue                      # pre-existing run: evaluate only, never retrain/overwrite
        train_seed(s, new_root)
        runs[label] = new_root

runs = {k: v for k, v in sorted(runs.items())
        if any((v / m / "checkpoints" / CKPT_NAME).exists() for m in MODEL_NAMES)}
if not runs:
    raise FileNotFoundError(
        "No checkpoints found. Looked for <root>/<Model>/checkpoints/" + CKPT_NAME +
        f" under: {[str(p) for p in SEARCH_PATHS]}.\n"
        "Fix: upload ALL_checkpoints.zip (or the three <Model>_checkpoints.zip files) to the "
        "working directory, or set CKPT_ROOTS = ['/path/to/folder_with_PointNet_etc'].")
print("\nRuns used for evaluation:", {k: str(v) for k, v in runs.items()})

# ---- 3. test sets + r_rms ----
TEST = load_or_build_test_sets()
rms_df = pd.DataFrame([{"cell": n, "group": g, "sigma": nz, "rms_mean": TEST[i][2].mean(),
                        "rms_std": TEST[i][2].std(),
                        "effective_noise_approx": nz / TEST[i][2].mean()}
                       for i, (n, g, sp, nz) in enumerate(CELLS)])
rms_df.to_csv(OUT / "rms_per_cell.csv", index=False)
print("\nPre-normalisation RMS radius / approximate effective noise (sigma / r_rms):")
print(rms_df.round(4).to_string(index=False))

# ---- 4. evaluate every run ----
rows, abl_rows, info_rows = [], [], []
corr_store = defaultdict(list)
cm_pool = {}
for run, root in runs.items():
    for name in MODEL_NAMES:
        path = root / name / "checkpoints" / CKPT_NAME
        if not path.exists():
            print(f"[skip] {run}/{name}: {path} not found")
            continue
        model, ck = load_model(path, name)
        sp = root / name / "training_summary.json"
        summ = json.load(open(sp)) if sp.exists() else {}
        info_rows.append({"run": run, "model": name,
                          "params": sum(p.numel() for p in model.parameters()),
                          "ckpt_epoch": ck.get("epoch"),
                          "best_val_acc_indist": ck.get("best_val_accuracy"),
                          "actual_epochs": summ.get("actual_epochs"),
                          "train_seconds": summ.get("training_time_seconds"),
                          "final_alpha": float(model.alpha.item()) if name == "CPGMM" else None})
        print(f"\n=== {run} / {name} ===")

        if name == "CPGMM":     # sanity check: replicated forward == real forward
            xb = torch.from_numpy(TEST[0][0][:32]).to(DEVICE)
            with torch.no_grad():
                diff = float((model(xb)[0] - cpgmm_logits(model, xb, "full")).abs().max())
            print(f"  forward-replica max |diff| = {diff:.2e}", "" if diff < 1e-3 else "  <-- CHECK")

        cm = np.zeros((NUM_CLASSES, NUM_CLASSES), dtype=np.int64)
        for i, (cname, group, spec, noise) in enumerate(CELLS):
            X, y, _ = TEST[i]
            pred = predict(model, name, X)
            m, correct = metrics(pred, y)
            lo, hi = boot_ci(correct)
            rows.append({"run": run, "model": name, "cell": cname, "group": group, **m,
                         "acc12_ci_lo": lo * 100, "acc12_ci_hi": hi * 100})
            corr_store[(run, name, group)].append(correct)
            if group == "paper-OOD":
                cm += np.bincount(y * NUM_CLASSES + pred,
                                  minlength=NUM_CLASSES ** 2).reshape(NUM_CLASSES, NUM_CLASSES)
            print(f"  {cname:24s} acc12 = {m['acc12']:6.2f}  [{lo * 100:.2f}, {hi * 100:.2f}]", flush=True)
        cm_pool[(run, name)] = cm

        if name == "CPGMM":
            const_emb = mean_proto_embedding(model, TEST[0][0])
            for mode in ["full", "no_bias", "no_proto", "const_proto",
                         "zero_c", "zero_sigma", "shuffle_sig"]:
                for i, (cname, group, spec, noise) in enumerate(CELLS):
                    if group not in ABL_GROUPS:
                        continue
                    X, y, _ = TEST[i]
                    sub = stratified_idx(y, ABL_PER_CLASS, seed=i)
                    Xs, ys = X[sub], y[sub]
                    src = mismatched_source_idx(ys, seed=i) if mode == "shuffle_sig" else None
                    pred = predict(model, name, Xs, mode=mode, src_idx=src, const_emb=const_emb)
                    m, _ = metrics(pred, ys)
                    abl_rows.append({"run": run, "mode": mode, "cell": cname, "group": group, **m})
            print("  ablations done", flush=True)

if not rows:
    raise RuntimeError("Checkpoints were located but nothing could be evaluated.")

# ---- 5. save + aggregate ----
df = pd.DataFrame(rows)
df.to_csv(OUT / "results_per_cell.csv", index=False)
pd.DataFrame(info_rows).to_csv(OUT / "checkpoint_info.csv", index=False)

grp_rows = []
for (run, name, group), arrs in corr_store.items():
    c = np.concatenate(arrs)
    lo, hi = boot_ci(c)
    grp_rows.append({"run": run, "model": name, "group": group, "acc12": c.mean() * 100,
                     "ci_lo": lo * 100, "ci_hi": hi * 100, "n_clouds": len(c)})
grp_df = pd.DataFrame(grp_rows)
grp_df.to_csv(OUT / "results_per_group.csv", index=False)

for (run, name), cm in cm_pool.items():
    pd.DataFrame(cm, index=CLASS_NAMES, columns=CLASS_NAMES).to_csv(
        OUT / f"confusion_paperOOD_{run}_{name}.csv")

GROUPS = ["in-dist", "noise-only", "shear-only", "paper-OOD"]
n_runs = grp_df["run"].nunique()
print("\n" + "=" * 78)
print(f"RESULTS ACROSS {n_runs} RUN(S)  (mean ± std over seeds; CIs in CSVs are test-sampling only)")
print("=" * 78)

tabA = pd.DataFrame({g: {m: pm(grp_df[(grp_df.model == m) & (grp_df["group"] == g)]["acc12"])
                         for m in MODEL_NAMES if m in set(grp_df.model)} for g in GROUPS})
print("\n12-class accuracy (%) by test group:")
print(tabA.to_string())
tabA.to_csv(OUT / "table_A_group_accuracy.csv")

paper = df[df["group"] == "paper-OOD"]
per_run = paper.groupby(["run", "model"])[["acc_2D", "acc_3D", "acc_4D", "acc_5D",
                                           "acc12", "acc_dim", "acc_family"]].mean().reset_index()
tabB = pd.DataFrame({c: {m: pm(per_run[per_run.model == m][c]) for m in MODEL_NAMES
                         if m in set(per_run.model)}
                     for c in ["acc_2D", "acc_3D", "acc_4D", "acc_5D", "acc12", "acc_dim", "acc_family"]})
print("\nTable-1 style, paper-OOD cells (%):")
print(tabB.to_string())
tabB.to_csv(OUT / "table_B_paperOOD_by_dim.csv")

piv = grp_df.pivot_table(index=["run", "group"], columns="model", values="acc12")
paired = {}
for base in ("PointNet", "VanillaTransformer"):
    if "CPGMM" in piv.columns and base in piv.columns:
        d = (piv["CPGMM"] - piv[base]).dropna()
        paired[f"CPGMM - {base}"] = {
            g: (pm(d.xs(g, level="group").values) + f", wins {int((d.xs(g, level='group') > 0).sum())}/"
                f"{len(d.xs(g, level='group'))}") for g in GROUPS if g in d.index.get_level_values("group")}
if paired:
    tabC = pd.DataFrame(paired).T
    print("\nPaired difference in 12-class accuracy (pp) per seed:")
    print(tabC.to_string())
    tabC.to_csv(OUT / "table_C_paired_differences.csv")

if abl_rows:
    abl = pd.DataFrame(abl_rows)
    abl.to_csv(OUT / "cpgmm_ablation_per_cell.csv", index=False)
    ag = abl.groupby(["run", "mode", "group"])["acc12"].mean().reset_index()
    tabD = pd.DataFrame({g: {md: pm(ag[(ag["mode"] == md) & (ag["group"] == g)]["acc12"])
                             for md in ag["mode"].unique()} for g in ABL_GROUPS})
    print(f"\nCP-GMM inference-time ablations (12-class %, stratified subset of {ABL_PER_CLASS}/class):")
    print(tabD.to_string())
    tabD.to_csv(OUT / "table_D_cpgmm_ablations.csv")

print("\nCheckpoint info:")
print(pd.DataFrame(info_rows).to_string(index=False))
print(f"\nAll files in: {OUT.resolve()}   (total {(time.time() - t_start) / 60:.1f} min)")
```

```python
# ===== CPU-ONLY: reload saved results + attention-bias diagnostic =====
from google.colab import drive
drive.mount('/content/drive')
import math, numpy as np, pandas as pd, torch, torch.nn as nn, torch.nn.functional as F
from pathlib import Path

WORK = Path("/content/drive/MyDrive/CPGMM")
OUT = WORK / "eval_results"
ROOTS = {"seed42": WORK,
         "seed43": WORK.parent / "quadratic_5D_experiment_seed43",
         "seed44": WORK.parent / "quadratic_5D_experiment_seed44"}

# ---------- PART 1: tables from saved CSVs (pandas only) ----------
GROUPS = ["in-dist", "noise-only", "shear-only", "paper-OOD"]
MODELS = ["PointNet", "VanillaTransformer", "CPGMM"]
def pm(x):
    x = np.asarray(x, float)
    return f"{x.mean():.2f} ± {x.std(ddof=1):.2f}" if len(x) > 1 else f"{x.mean():.2f}"

df = pd.read_csv(OUT / "results_per_cell.csv")
gm = df.groupby(["run", "model", "group"])["acc12"].mean().reset_index()
print("12-class accuracy (%), mean ± std over 3 seeds:")
print(pd.DataFrame({g: {m: pm(gm[(gm["model"] == m) & (gm["group"] == g)]["acc12"])
                        for m in MODELS} for g in GROUPS}).to_string())

cols = ["acc_2D", "acc_3D", "acc_4D", "acc_5D", "acc12", "acc_dim", "acc_family"]
paper = df[df["group"] == "paper-OOD"]
print("\nPer-seed, per-dimension accuracy on paper-OOD cells (%):")
print(paper.groupby(["run", "model"])[cols].mean().round(1).to_string())

print("\nPer-dimension, mean ± std over seeds:")
pr = paper.groupby(["run", "model"])[cols].mean().reset_index()
print(pd.DataFrame({c: {m: pm(pr[pr["model"] == m][c]) for m in MODELS} for c in cols}).to_string())

abl = pd.read_csv(OUT / "cpgmm_ablation_per_cell.csv")
ag = abl.groupby(["run", "mode", "group"])["acc12"].mean().reset_index()
print("\nCP-GMM inference-time ablations (12-class %):")
print(pd.DataFrame({g: {md: pm(ag[(ag["mode"] == md) & (ag["group"] == g)]["acc12"])
                        for md in ag["mode"].unique()} for g in ["in-dist", "paper-OOD"]}).to_string())

# ---------- PART 2: attention-bias diagnostic (CPU, CP-GMM only) ----------
def quadratic_veronese(X):
    B, N, D = X.shape
    t = [X[:, :, i] * X[:, :, j] for i in range(D) for j in range(i, D)]
    t += [X[:, :, i] for i in range(D)] + [torch.ones_like(X[:, :, 0])]
    return torch.stack(t, dim=-1)

def coef_mat(c):
    A = torch.zeros(c.shape[0], 5, 5, dtype=c.dtype); k = 0
    for i in range(5):
        for j in range(i, 5):
            if i == j: A[:, i, j] = c[:, k]
            else: A[:, i, j] = A[:, j, i] = c[:, k] / 2.0
            k += 1
    return A

class CPAttentionBlock(nn.Module):
    def __init__(self, dim=128, heads=8):
        super().__init__()
        self.dim, self.heads, self.head_dim = dim, heads, dim // heads
        self.norm1 = nn.LayerNorm(dim); self.qkv = nn.Linear(dim, dim * 3)
        self.projection = nn.Linear(dim, dim); self.norm2 = nn.LayerNorm(dim)
        self.ffn = nn.Sequential(nn.Linear(dim, dim * 4), nn.GELU(), nn.Linear(dim * 4, dim))
    def forward(self, X, bias, alpha):
        B, N, D = X.shape
        q, k, v = self.qkv(self.norm1(X)).reshape(B, N, 3, self.heads, self.head_dim).permute(2, 0, 3, 1, 4)
        att = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_dim) + alpha * bias[:, None]
        X = X + self.projection((torch.softmax(att, -1) @ v).transpose(1, 2).reshape(B, N, D))
        return X + self.ffn(self.norm2(X))

class CPGMMNetwork(nn.Module):
    def __init__(self, nc=12):
        super().__init__()
        self.embedding = nn.Linear(26, 128)
        self.router = nn.Sequential(nn.Linear(42, 128), nn.LayerNorm(128), nn.GELU(), nn.Linear(128, 64), nn.GELU())
        self.templates = nn.Parameter(torch.randn(nc, 64) * 0.02)
        self.router_temperature = 1.0
        self.template_projection = nn.Linear(64, 128)
        self.blocks = nn.ModuleList([CPAttentionBlock(128, 8) for _ in range(3)])
        self.alpha = nn.Parameter(torch.tensor(1e-3))
        self.norm = nn.LayerNorm(128)
        self.classifier = nn.Sequential(nn.Linear(128, 256), nn.GELU(), nn.Dropout(0.3), nn.Linear(256, nc))
    def algebraic_features(self, X):
        with torch.no_grad():
            lifted = quadratic_veronese(X); L = lifted.double()
            ev, V = torch.linalg.eigh(L.transpose(1, 2) @ L)
            c = V[:, :, 0].float()
            c = c * torch.where(c[:, 0:1] < 0, -torch.ones_like(c[:, 0:1]), torch.ones_like(c[:, 0:1]))
            s = torch.sqrt(torch.clamp(ev.flip(dims=[1]), min=0.0)).float()
            s = s / (s.norm(dim=1, keepdim=True) + 1e-8)
            return lifted, c, torch.cat([c, s], 1)

@torch.no_grad()
def diag(model, X_np, y_np, label, per_class=50):
    rng = np.random.default_rng(0)
    sub = np.sort(np.concatenate([rng.permutation(np.where(y_np == c)[0])[:per_class] for c in range(12)]))
    X, y = torch.from_numpy(X_np[sub]), y_np[sub]
    lifted, coef, alg = model.algebraic_features(X)
    Bm = (X @ coef_mat(coef) @ X.transpose(1, 2)) / math.sqrt(21)
    alpha = model.alpha.item()
    nr, nt = F.normalize(model.router(alg), dim=1), F.normalize(model.templates, dim=1)
    w = torch.softmax(nr @ nt.T / model.router_temperature, -1)
    tok = model.embedding(torch.cat([X, lifted], -1)) + model.template_projection(w @ model.templates)[:, None, :]
    blk = model.blocks[0]; Z = blk.norm1(tok); B_, N, _ = Z.shape
    q, k, _ = blk.qkv(Z).reshape(B_, N, 3, blk.heads, blk.head_dim).permute(2, 0, 3, 1, 4)
    qk = (q @ k.transpose(-2, -1)) / math.sqrt(blk.head_dim)
    def run(a):
        t = tok
        for b in model.blocks: t = b(t, Bm, a)
        return model.classifier(model.norm(t.mean(1)))
    full, nob = run(alpha), run(0.0)
    qs, bs = qk.std(-1).mean().item(), (alpha * Bm).std(-1).mean().item()
    print(f"[{label}] alpha={alpha:.3f} | bias std={Bm.std().item():.2e} | alpha*bias spread={bs:.2e} "
          f"vs QK-logit spread={qs:.2e} (ratio {bs / qs:.2e})")
    print(f"      max|dlogit|={(full - nob).abs().max().item():.2e} | prediction flips "
          f"{int((full.argmax(1) != nob.argmax(1)).sum())}/{len(sub)}")
    print("      bias std by span-dim:", ", ".join(
        f"{d}D {Bm[torch.from_numpy(y // 3 == i)].std().item():.2e}" for i, d in enumerate([2, 3, 4, 5])))

z = np.load(OUT / "test_sets_seed2024_n300.npz")
for run, root in ROOTS.items():
    m = CPGMMNetwork()
    m.load_state_dict(torch.load(root / "CPGMM/checkpoints/best_validation.pt",
                                 map_location="cpu", weights_only=False)["model_state_dict"])
    m.eval()
    for ci, name in [(0, "in-dist"), (10, "paper s=30")]:   # cell 0 = in-dist, cell 10 = paper s=30
        diag(m, z[f"X{ci}"], z[f"y{ci}"], f"{run} | {name}")
```

```python
# ===== CPU-ONLY: redraw Figure 1, rebuild Tables 1-3 from saved results =====
from google.colab import drive
drive.mount('/content/drive')
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from pathlib import Path

OUT = Path("/content/drive/MyDrive/CPGMM/eval_results")
MODELS = ["PointNet", "VanillaTransformer", "CPGMM"]
LABEL = {"PointNet": "PointNet", "VanillaTransformer": "Vanilla Transformer", "CPGMM": "CP-GMM (ours)"}
COLOR = {"PointNet": "#1f77b4", "VanillaTransformer": "#ff7f0e", "CPGMM": "#2ca02c"}
DIMS = [2, 3, 4, 5]
FAMS = ["Ellipsoid", "Hyperboloid", "Paraboloid"]
CLASS = [f"{d}D_{f}" for d in DIMS for f in FAMS]

def pm(x):
    x = np.asarray(x, float)
    return f"{x.mean():.2f} ± {x.std(ddof=1):.2f}" if len(x) > 1 else f"{x.mean():.2f}"

df = pd.read_csv(OUT / "results_per_cell.csv")
df = df[df["model"].isin(MODELS)]
runs = sorted(df["run"].unique())
paper = df[df["group"] == "paper-OOD"].copy()
paper["s"] = paper["cell"].str.extract(r"_s(\d+)_")[0].astype(int)
print("Runs:", runs)

# ---------------- FIGURE 1 ----------------
dcols = [f"acc_{d}D" for d in DIMS]
per_run_dim = paper.groupby(["run", "model"])[dcols].mean().reset_index()
stat_dim = per_run_dim.groupby("model")[dcols].agg(["mean", "std"])
per_run_s = paper.groupby(["run", "model", "s"])["acc12"].mean().reset_index()
stat_s = per_run_s.groupby(["model", "s"])["acc12"].agg(["mean", "std"]).reset_index()

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
w = 0.26
for k, m in enumerate(MODELS):
    means = [stat_dim.loc[m, (c, "mean")] for c in dcols]
    stds = [np.nan_to_num(stat_dim.loc[m, (c, "std")]) for c in dcols]
    ax[0].bar(np.arange(4) + (k - 1) * w, means, w, yerr=stds, capsize=3,
              color=COLOR[m], label=LABEL[m])
ax[0].set_xticks(range(4)); ax[0].set_xticklabels([f"{d}D" for d in DIMS])
ax[0].set_ylim(0, 100); ax[0].set_xlabel("Span dimension d"); ax[0].set_ylabel("Paper-OOD accuracy (%)")
ax[0].set_title(f"(a) Accuracy by span dimension (mean ± std, {len(runs)} seeds)", fontsize=10)
ax[0].legend(fontsize=8)
for m in MODELS:
    t = stat_s[stat_s["model"] == m].sort_values("s")
    ax[1].errorbar(t["s"], t["mean"], yerr=t["std"].fillna(0), marker="o", capsize=3,
                   color=COLOR[m], label=LABEL[m])
ax[1].axvspan(0.5, 2.5, color="gray", alpha=0.15)
ax[1].text(1.0, 12, "training\nshear range", fontsize=7, ha="center")
ax[1].set_ylim(0, 100); ax[1].set_xlabel("Shear bound s (test noise σ=0.10)")
ax[1].set_ylabel("12-class accuracy (%)")
ax[1].set_title("(b) Accuracy vs shear (paper-OOD cells)", fontsize=10)
plt.tight_layout()
plt.savefig(OUT / "figure1_updated.png", dpi=300); plt.savefig(OUT / "figure1_updated.pdf")
plt.show()

# Figure 2: separated shifts
fig, ax = plt.subplots(figsize=(6.5, 3.6))
gm = df.groupby(["run", "model", "group"])["acc12"].mean().reset_index()
G = ["in-dist", "noise-only", "shear-only", "paper-OOD"]
for k, m in enumerate(MODELS):
    mm = [gm[(gm["model"] == m) & (gm["group"] == g)]["acc12"] for g in G]
    ax.bar(np.arange(4) + (k - 1) * w, [x.mean() for x in mm],
           w, yerr=[np.nan_to_num(x.std(ddof=1)) for x in mm], capsize=3,
           color=COLOR[m], label=LABEL[m])
ax.set_xticks(range(4)); ax.set_xticklabels(["In-dist", "Noise-only", "Shear-only\n(s≥5)", "Paper OOD"])
ax.set_ylim(0, 100); ax.set_ylabel("12-class accuracy (%)"); ax.legend(fontsize=8)
ax.set_title("Separated shifts (mean ± std over seeds)", fontsize=10)
plt.tight_layout(); plt.savefig(OUT / "figure2_separated_shifts.png", dpi=300); plt.show()

# ---------------- TABLE 1 ----------------
cols = dcols + ["acc12", "acc_dim", "acc_family"]
pr = paper.groupby(["run", "model"])[cols].mean().reset_index()
t1 = pd.DataFrame({c: {LABEL[m]: pm(pr[pr["model"] == m][c]) for m in MODELS} for c in cols})
t1.columns = ["2D", "3D", "4D", "5D", "12-class", "Dim-only", "Family-only"]
print("\nTABLE 1 (paper-OOD, mean ± std over seeds):"); print(t1.to_string())
t1.to_csv(OUT / "table1_updated.csv")

# ---------------- TABLE 2 (CP-GMM ablations by shear) ----------------
abl = pd.read_csv(OUT / "cpgmm_ablation_per_cell.csv")
abl = abl[abl["group"] == "paper-OOD"].copy()
abl["s"] = abl["cell"].str.extract(r"_s(\d+)_")[0].astype(int)
order = ["full", "no_bias", "no_proto", "const_proto", "shuffle_sig", "zero_c", "zero_sigma"]
t2 = {}
for md in order:
    row = {f"s={s}": pm(abl[(abl["mode"] == md) & (abl["s"] == s)]["acc12"]) for s in (1, 5, 12, 20, 30)}
    row["Mean"] = pm(abl[abl["mode"] == md].groupby("run")["acc12"].mean())
    t2[md] = row
t2 = pd.DataFrame(t2).T
print("\nTABLE 2 (inference-time ablations, 100 clouds/class subset, mean ± std over seeds):")
print(t2.to_string()); t2.to_csv(OUT / "table2_updated.csv")

# ---------------- TABLE 3 (worst family + dominant error transition) ----------------
rows, share_rows = [], []
for m in MODELS:
    mats = []
    for run in runs:
        p = OUT / f"confusion_paperOOD_{run}_{m}.csv"
        if p.exists():
            cm = pd.read_csv(p, index_col=0).loc[CLASS, CLASS].values.astype(float)
            mats.append(cm / cm.sum(1, keepdims=True))
    if not mats:
        continue
    R = np.mean(mats, 0)
    for di, d in enumerate(DIMS):
        idx = list(range(3 * di, 3 * di + 3))
        acc = {CLASS[i]: R[i, i] * 100 for i in idx}
        worst = min(acc, key=acc.get); wi = CLASS.index(worst)
        wvals = [mm[wi, wi] * 100 for mm in mats]
        off = R.copy(); np.fill_diagonal(off, 0)
        i, j = np.unravel_index(off[idx].argmax(), off[idx].shape)
        rows.append({"Dim": f"{d}D", "Model": LABEL[m], "Worst shape": worst.split("_")[1],
                     "Acc (%)": pm(wvals),
                     "Dominant error": f"{CLASS[idx[i]]} -> {CLASS[j]} ({off[idx[i], j] * 100:.1f}%)"})
    # where do errors go? (rows = true class; classes equally sized)
    err = R.copy(); np.fill_diagonal(err, 0); tot = err.sum()
    fam = np.array([c % 3 for c in range(12)]); dim = np.array([c // 3 for c in range(12)])
    same_f = sum(err[a, b] for a in range(12) for b in range(12) if fam[a] == fam[b] and dim[a] != dim[b])
    same_d = sum(err[a, b] for a in range(12) for b in range(12) if dim[a] == dim[b] and fam[a] != fam[b])
    both = tot - same_f - same_d
    share_rows.append({"Model": LABEL[m], "same family, other dim": f"{same_f / tot * 100:.1f}%",
                       "same dim, other family": f"{same_d / tot * 100:.1f}%",
                       "both differ": f"{both / tot * 100:.1f}%"})
t3 = pd.DataFrame(rows)
print("\nTABLE 3 (mean over seeds; worst-family accuracy ± std over seeds):")
print(t3.to_string(index=False)); t3.to_csv(OUT / "table3_updated.csv", index=False)
print("\nShare of ALL errors by error type (paper-OOD):")
print(pd.DataFrame(share_rows).to_string(index=False))
print("\nSaved figure1_updated.png/.pdf, figure2_separated_shifts.png, table1-3_updated.csv in", OUT)
```
